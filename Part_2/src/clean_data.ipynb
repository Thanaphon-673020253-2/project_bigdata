```python
import pandas as pd
import ast

df = pd.read_csv(f"W:\Learn_Program\project_bigdata\Part_2\data\data.csv")

def parse_categories(x):
    if pd.isna(x):
        return []
    
    try:
        return ast.literal_eval(x)
    except:
        return []

df["categories"] = df["categories"].apply(parse_categories)

# เอาเฉพาะข้อมูลสินค้าที่ไม่ซ้ำ
products = (
    df[
        [
            "asin",
            "parent_asin",
            "title_product",
            "categories"
        ]
    ]
    .drop_duplicates("asin")
    .reset_index(drop=True)
)

display(products.head(10))
```

```python
import pandas as pd
import numpy as np
import ast

# ตรวจสอบก่อน
print(df["categories"].apply(type).value_counts())
```

```python
def get_category_level(categories, level):
    if isinstance(categories, list) and len(categories) >= level:
        return categories[level - 1]
    return None


for level in range(1, 6):
    df[f"category_level_{level}"] = df["categories"].apply(
        lambda x: get_category_level(x, level)
    )
```
