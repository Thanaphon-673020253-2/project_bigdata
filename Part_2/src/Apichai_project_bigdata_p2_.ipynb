### 2.1 เลือก Client Brief

เลือก **1 brief** จากรายการนี้ หรือเขียนเองก็ได้ถ้าคิดโจทย์ที่ดีกว่าออก (ส่งให้อาจารย์อนุมัติก่อน)
สิ่งที่ทำให้ brief ใช้ได้คือ **มีคนที่ต้องตัดสินใจ และการตัดสินใจนั้นมีต้นทุน**

| Brief | ลูกค้าคือทีม | คำถามที่เขาต้องการคำตอบ |
|---|---|---|
| **B1 ปรับปรุงสินค้า** | Product | สินค้ากลุ่มไหนควรแก้ก่อน และต้องแก้เรื่องอะไร งบจำกัดแก้ได้แค่ 3 เรื่อง |
| **B2 ปฏิบัติการและการจัดส่ง** | Operations | ปัญหาบรรจุภัณฑ์และการจัดส่งกินคะแนนไปเท่าไร คุ้มไหมที่จะลงทุนเปลี่ยนแพ็กเกจ |
| **B3 สร้างข้อความการตลาด** | Marketing | ลูกค้าที่พอใจใช้คำว่าอะไร เอาคำของลูกค้าจริงมาทำ copy และ claim อะไรได้บ้าง |
| **B4 ความน่าเชื่อถือของรีวิว** | Trust and Safety | รีวิวที่น่าสงสัยว่าไม่ได้มาจากการใช้จริง มีลักษณะทางภาษาแบบใด ควรตั้งกฎคัดกรองอย่างไร |
| **B5 การจัดอันดับและการคัดสินค้า** | Marketplace | ควรใช้อะไรเป็นสัญญาณว่ารีวิวไหน "มีประโยชน์" และสินค้าแบบไหนควรลดการโปรโมท |
| **B6 ราคาและความคาดหวัง** | Pricing | ลูกค้าที่ซื้อสินค้าราคาสูงกับราคาต่ำ คาดหวังต่างกันอย่างไร สะท้อนในภาษาที่เขาใช้อย่างไร |

**<Brief ที่ทีมเลือก>**

* ลูกค้าคือทีม: **ทีมบริหารและพัฒนาผลิตภัณฑ์ (Product Management & Category Team)**
* เขากำลังจะตัดสินใจเรื่อง: **การจัดสรรงบประมาณ R&D และแก้ไขปรับปรุงสินค้า (Product Improvement) โดยต้องเลือกหมวดหมู่สินค้าที่วิกฤตที่สุด และเลือกแก้ไขเพียง 3 ปัญหาหลัก ที่ส่งผลกระทบต่อคะแนนรีวิวมากที่สุดเนื่องจากงบประมาณจำกัด**
* ถ้าตัดสินใจผิดจะเสียอะไร: **สูญเสียงบประมาณ R&D ไปกับการแก้ปัญหาที่ไม่ตรงจุด (Wasted Budget) ในขณะที่ปัญหาจริงยังคงอยู่ ส่งผลให้สินค้าถูกลดความน่าเชื่อถือ คะแนนรีวิวเฉลี่ยตก เสียส่วนแบ่งทางการตลาดให้คู่แข่ง และสูญเสียรายได้ในระยะยาว**
* คำถามที่เราจะตอบให้เขา:
**1. สินค้าหมวดหมู่ใดใน Office Products ที่มีสัดส่วนรีวิวเชิงลบ (1–2 ดาว) สูงที่สุด และควรเร่งแก้ไขก่อนเป็นอันดับแรก**
**2. 3 ปัญหาหลัก (Root Causes) ที่ลูกค้าบ่นถึงมากที่สุดในรีวิวเชิงลบของสินค้าหมวดหมู่นั้นคืออะไร (เช่น ความคงทนของวัสดุ, ชิ้นส่วนประกอบไม่ครบ/ชำรุด, หรือการใช้งานไม่ตรงตามสเปก)**
**3. หากแก้ไข 3 ปัญหานี้สำเร็จ จะมีโอกาสปรับเพิ่มคะแนนรีวิวเฉลี่ยของหมวดหมู่นั้นได้สูงสุดกี่ดาว**
* สมมติฐานตั้งต้นของเรา (จะถูกหรือผิดก็ได้):  **สินค้ากลุ่มที่มีชิ้นส่วนกลไกหรือโครงสร้างขนาดใหญ่ เช่น Printers & Accessories, Chairs & Sofas และ Carts & Stands มีอัตราส่วนรีวิวเชิงลบ (1–2 ดาว) สูงกว่ากลุ่มเครื่องเขียนทั่วไป โดยปัญหาหลักเกิน 50% เกิดจากความคงทนของวัสดุและการประกอบ ไม่ใช่เรื่องดีไซน์หรือรูปลักษณ์**

### 2.2 เตรียมข้อมูลที่เชื่อถือได้

ทำเพียงสิ่งจำเป็นก่อนเริ่มวิเคราะห์

````python
"""
หากยังไม่มี library ใด ให้ติดตั้งก่อน
!pip install -q nltk wordcloud scikit-learn pandas matplotlib tqdm huggingface_hub
!pip install -q pythainlp          # ใช้เมื่อทำงานกับข้อความภาษาไทย
"""
import json, os, re, random, itertools, time
from collections import Counter, defaultdict

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

import nltk
for pkg in ["punkt_tab", "stopwords", "wordnet", "vader_lexicon"]:
    nltk.download(pkg, quiet=True)

%matplotlib inline

RANDOM_SEED = 42
SAMPLE_N = 20_000         # เวอร์ชันลด workload: Part 2 ใช้อย่างน้อย 20,000 รีวิว
random.seed(RANDOM_SEED)
np.random.seed(RANDOM_SEED)

DATA_DIR = "./data/"
os.makedirs(DATA_DIR, exist_ok=True)
````

````python
# ดาวน์โหลดไฟล์เฉพาะหมวดที่เลือก
from huggingface_hub import hf_hub_download

CATEGORY = "Office_Products"   # <-- เปลี่ยนเป็นหมวดที่ทีมจอง

review_path = hf_hub_download(repo_id="McAuley-Lab/Amazon-Reviews-2023",
                              filename=f"raw/review_categories/{CATEGORY}.jsonl",
                              repo_type="dataset", local_dir=DATA_DIR)
meta_path   = hf_hub_download(repo_id="McAuley-Lab/Amazon-Reviews-2023",
                              filename=f"raw/meta_categories/meta_{CATEGORY}.jsonl",
                              repo_type="dataset", local_dir=DATA_DIR)
print(review_path, meta_path)
````

````python
from tqdm import tqdm

# อ่านไฟล์ review_path ทีละบรรทัดจนครบ SAMPLE_N
reviews = []
with open(review_path, "r", encoding="utf-8") as f:
    for i, line in enumerate(tqdm(f, desc="Loading Reviews")):
        reviews.append(json.loads(line))
        if len(reviews) >= SAMPLE_N:
            break

df_reviews = pd.DataFrame(reviews)
print(f"✅ โหลดข้อมูลรีวิวสำเร็จ: {df_reviews.shape[0]:,} รายการ")
````

````python
from tqdm import tqdm

# 1. เก็บ parent_asin ทั้งหมดจาก 20,000 รีวิวไว้ใน Set เพื่อการค้นหาที่รวดเร็ว
target_asins = set(df_reviews["parent_asin"].dropna().unique())

# 2. อ่านไฟล์ meta_path และดึงเฉพาะรายการที่ตรงกับ target_asins
meta_data = []
with open(meta_path, "r", encoding="utf-8") as f:
    for line in tqdm(f, desc="Loading Meta Data"):
        data = json.loads(line)
        if data.get("parent_asin") in target_asins:
            meta_data.append(data)

df_meta = pd.DataFrame(meta_data)
print(f"✅ โหลดข้อมูล Metadata สำเร็จ: {df_meta.shape[0]:,} รายการ")
````

````python
# 1. เช็กจำนวนสินค้าที่ไม่ซ้ำในฝั่งรีวิว
print("จำนวนสินค้าที่ไม่ซ้ำในรีวิว:", df_reviews["parent_asin"].nunique())

# 2. ทำการ Merge ข้อมูล
df_merged = pd.merge(
    df_reviews,
    df_meta,
    on="parent_asin",
    how="left",
    suffixes=("_review", "_product")
)

# 3. เช็กจำนวนแถวผลลัพธ์
print(f"✅ จำนวนบรรทัดหลัง Merge: {df_merged.shape[0]:,} รายการ")
df_merged[["rating", "parent_asin", "title_product", "text"]].head(3)
````

````python
import pandas as pd
df_reviews = pd.read_csv(r"W:\Learn_Program\project_bigdata\Part_2\data\data.csv")
````

````python
# สร้างตัวแปรสำหรับระบุรีวิวเชิงลบ
df_reviews["negative_review"] = df_reviews["rating"] <= 2

# สรุปจำนวนและสัดส่วนรีวิวเชิงลบ
negative_summary = pd.DataFrame({
    "จำนวนรีวิว": [
        len(df_reviews),
        df_reviews["negative_review"].sum()
    ],
    "ร้อยละ": [
        100,
        df_reviews["negative_review"].mean() * 100
    ]
}, index=["รีวิวทั้งหมด", "รีวิวเชิงลบ (1–2 ดาว)"])

display(negative_summary)
````

````python
import pandas as pd
import matplotlib.pyplot as plt

# สรุปจำนวนและร้อยละของแต่ละคะแนน
rating_summary = (
    df_reviews["rating"]
    .value_counts()
    .sort_index()
    .rename("จำนวนรีวิว")
    .to_frame()
)

rating_summary["ร้อยละ"] = (
    rating_summary["จำนวนรีวิว"] / len(df_reviews) * 100
)

display(rating_summary)

````

**จากรีวิวทั้งหมด 20,000 รายการ พบว่ารีวิว 5 ดาวมีสัดส่วนสูงที่สุด 72.59% ขณะที่รีวิวเชิงลบระดับ 1–2 ดาวมีจำนวนรวม 1,540 รายการ หรือ 7.70% ของข้อมูลทั้งหมด แม้จะเป็นสัดส่วนที่น้อยกว่ารีวิวเชิงบวก แต่เป็นกลุ่มที่สะท้อนปัญหาและความไม่พึงพอใจของลูกค้า ผู้พัฒนาจึงมองเห็นว่าควรนำไปวิเคราะห์ต่อเพื่อค้นหาสาเหตุของปัญหาที่เกิดขึ้นกับสินค้า**

````python
# สร้างตัวแปรรีวิวเชิงลบ
df_reviews["negative_review"] = df_reviews["rating"] <= 2

# คำนวณสัดส่วนรีวิวเชิงลบตามสถานะ Verified Purchase
verified_summary = (
    df_reviews
    .groupby("verified_purchase")["negative_review"]
    .agg(["count", "sum", "mean"])
    .reset_index()
)

verified_summary["negative_rate_percent"] = (
    verified_summary["mean"] * 100
)

verified_summary = verified_summary[
    ["verified_purchase", "count", "sum", "negative_rate_percent"]
]

verified_summary.columns = [
    "Verified Purchase",
    "จำนวนรีวิว",
    "รีวิวเชิงลบ (1–2 ดาว)",
    "สัดส่วนรีวิวเชิงลบ (%)"
]

display(verified_summary)
````

**เมื่อพิจารณาสถานะ Verified Purchase พบว่า กลุ่มผู้ซื้อที่ยืนยันการซื้อมีสัดส่วนรีวิวเชิงลบระดับ 1–2 ดาวประมาณ 9.45% ขณะที่กลุ่มที่ไม่ได้ยืนยันการซื้อมีสัดส่วนประมาณ 3.77% จึงพบว่ากลุ่ม Verified Purchase มีสัดส่วนรีวิวเชิงลบสูงกว่า อย่างไรก็ตาม ผลดังกล่าวเป็นเพียงความแตกต่างที่พบจากข้อมูลเบื้องต้น และยังไม่สามารถสรุปได้ว่าสถานะ Verified Purchase เป็นสาเหตุโดยตรงที่ทำให้เกิดคะแนนรีวิวต่ำ**

````python
import matplotlib.pyplot as plt

# 1. เตรียมข้อมูลโดยจัดกลุ่มนับจำนวนรีวิวตาม Rating และ Verified Purchase
grouped_data = (
    df_reviews
    .groupby(["rating", "verified_purchase"])
    .size()
    .unstack(fill_value=0)
)

# 2. วาดกราฟแท่งแบบจัดกลุ่ม (Grouped Bar Chart)
# สมมติว่า verified_purchase มีค่า False (ซ้าย) และ True (ขวา) จะใช้สีแดงและเขียวตามลำดับ
ax = grouped_data.plot(
    kind="bar",
    figsize=(9, 5),
    width=0.7,
    color=["#d62728", "#2ca02c"]
)

plt.xlabel("Rating", fontsize=12)
plt.ylabel("Number of Reviews", fontsize=12)
plt.title("Distribution of Ratings by Verified Purchase", fontsize=14, pad=15)
plt.xticks(rotation=0) # ให้ตัวเลขแกน X ตั้งตรง
plt.legend(title="Verified Purchase")

# 3. ใส่ตัวเลขกำกับบนแท่งกราฟเพื่อให้เห็นสัดส่วนชัดเจน
for p in ax.patches:
    height = p.get_height()
    if height > 0:
        ax.annotate(f'{int(height):,}',
                    (p.get_x() + p.get_width() / 2., height),
                    ha='center', va='bottom',
                    xytext=(0, 3),
                    textcoords='offset points',
                    fontsize=9)

plt.tight_layout()
plt.show()
````

````python
df = df_reviews
````

````python
import ast

# ดึงชื่อหมวดหมู่สุดท้ายจาก categories
def get_last_category(x):
    try:
        categories = ast.literal_eval(x)

        if isinstance(categories, list) and len(categories) > 0:
            return categories[-1]

    except:
        pass

    return None


df["product_category"] = df["categories"].apply(get_last_category)

# สร้างตัวแปรรีวิวเชิงลบ
df["negative_review"] = df["rating"] <= 2

# สรุปตามหมวดสินค้า
category_summary = (
    df.groupby("product_category")
      .agg(
          จํานวนรีวิว=("rating", "size"),
          รีวิวเชิงลบ=("negative_review", "sum"),
          Negative_Rate=("negative_review", "mean")
      )
      .reset_index()
)

# แปลงเป็น %
category_summary["Negative_Rate"] = (
    category_summary["Negative_Rate"] * 100
)

# ใช้เฉพาะหมวดที่มีอย่างน้อย 100 รีวิว
# เพื่อไม่ให้หมวดที่มีข้อมูลน้อยมากทำให้สัดส่วนสูงผิดปกติ
category_filtered = category_summary[
    category_summary["จํานวนรีวิว"] >= 100
].copy()

# เรียงจาก Negative Rate สูงไปต่ำ
category_top = (
    category_filtered
    .sort_values("Negative_Rate", ascending=False)
    .head(10)
)

display(category_top)
````

````python
plt.figure(figsize=(9,6))

plt.barh(
    category_top["product_category"],
    category_top["Negative_Rate"]
)

plt.xlabel("Negative Review Rate (%)")
plt.ylabel("Product Category")
plt.title("Top Product Categories by Negative Review Rate")

plt.gca().invert_yaxis()

plt.show()
````

**หมวดสินค้าที่มี Negative Review Rate สูงถูกคัดเลือกเป็นกลุ่มที่ควรนำไปวิเคราะห์ข้อความเชิงลบต่อ เพื่อค้นหา Root Cause ก่อนตัดสินใจด้านการจัดสรรงบประมาณ R&D จึงควรวิเคราะห์ข้อความของรีวิว 1–2 ดาวในหมวดที่มีสัดส่วนสูง เพื่อค้นหาว่าปัญหาหลักที่ลูกค้ากล่าวถึงคืออะไร และใช้เป็นข้อมูลประกอบการกำหนด 3 ปัญหาที่ควรแก้ไขต่อไป**

````python
helpful_summary = (
    df_reviews
    .groupby("rating")["helpful_vote"]
    .agg(["count", "mean", "median"])
    .reset_index()
)

helpful_summary.columns = [
    "Rating",
    "จำนวนรีวิว",
    "Average Helpful Vote",
    "Median Helpful Vote"
]

display(helpful_summary)
````

````python
plt.figure(figsize=(7,4))

plt.bar(
    helpful_summary["Rating"],
    helpful_summary["Average Helpful Vote"]
)

plt.xlabel("Rating")
plt.ylabel("Average Helpful Vote")
plt.title("Average Helpful Vote by Rating")
plt.xticks([1, 2, 3, 4, 5])

plt.show()
````

**รีวิวที่มีคะแนนต่ำมีค่า Average Helpful Vote สูงกว่ารีวิวที่มีคะแนนสูง โดยรีวิว 1 ดาวมีค่าเฉลี่ยประมาณ 2.18 ขณะที่รีวิว 5 ดาวมีค่าเฉลี่ยประมาณ 0.92 ลักษณะดังกล่าวทำให้รีวิวเชิงลบน่าสนใจสำหรับการวิเคราะห์ต่อ เพราะอาจจะมีเนื้อหารีวิวที่ช่วยสะท้อนปัญหาของสินค้าและมีประโยชน์ต่อการค้นหาสาเหตุของปัญหา**

### 2.4 เตรียมข้อความและสร้าง Text Features

ก่อนนำข้อความไปวิเคราะห์ จำเป็นต้องมีการ **preprocessing** เพื่อทำให้ข้อความอยู่ในรูปที่เหมาะสมกับเทคนิคที่จะใช้ อย่างไรก็ตาม ในงานนี้ไม่จำเป็นต้องทำ preprocessing ทุกขั้นตอน ให้เลือกทำ **เท่าที่จำเป็น** ตามวัตถุประสงค์ของการวิเคราะห์

ตัวอย่างเช่น การเปลี่ยนข้อความเป็นตัวพิมพ์เล็ก (lower case) การนำ URL หรือ HTML ออก การแบ่งข้อความเป็นคำ (tokenization) หรือการตัด stopwords เช่น `the`, `and`, `is` ออก ในกรณีที่คำเหล่านี้ไม่ได้ช่วยในการวิเคราะห์

**ไม่บังคับให้ใช้ stemming หรือ lemmatization** หากไม่มีเหตุผลว่าขั้นตอนดังกล่าวจะช่วยตอบคำถามของทีมได้ดีขึ้น

หลังจากเตรียมข้อความแล้ว ให้สร้าง **Text Features อย่างน้อย 2 ชนิด** เพื่อนำข้อความมาเปลี่ยนให้อยู่ในรูปข้อมูลที่สามารถนำไปสำรวจ เปรียบเทียบ หรือวิเคราะห์ต่อได้

---

#### Text Feature คืออะไร?

Text Feature คือ **ข้อมูลเชิงตัวเลขที่สกัดหรือสร้างขึ้นจากข้อความ** เพื่อแทนลักษณะหรือความหมายบางอย่างของข้อความนั้น

Text Feature อาจอยู่ในรูปของ

* **ค่าตัวเลขค่าเดียว** เช่น จำนวนคำ หรือ sentiment score
* **ตัวแปรหลายตัว** เช่น `price=1`, `quality=0`, `delivery=1`
* **เวกเตอร์หลายมิติ** เช่น TF-IDF vector หรือ Text Embedding

ตัวอย่างเช่น รีวิว

> `"The product is easy to use but the delivery was very slow."`

สามารถเปลี่ยนเป็น feature ได้หลายรูปแบบ

| Feature                    |                 ตัวอย่างค่าที่ได้ |
| -------------------------- | --------------------------------: |
| จำนวนคำในรีวิว             |                                12 |
| Sentiment score            |                             -0.15 |
| กล่าวถึง usability หรือไม่ |                                 1 |
| กล่าวถึง delivery หรือไม่  |                                 1 |
| กล่าวถึง price หรือไม่     |                                 0 |
| TF-IDF vector              |   `[0.00, 0.41, 0.00, 0.72, ...]` |
| Embedding vector           | `[0.21, -0.08, 0.63, ..., -0.14]` |

ดังนั้นข้อความหนึ่งข้อความสามารถถูกแทนได้ทั้งด้วย **feature ที่ตีความได้โดยตรง** และ **feature แบบเวกเตอร์ที่ใช้แทนเนื้อหาหรือความหมายของข้อความ**

---

## ตัวอย่าง Text Features ที่สามารถเลือกใช้

ผมจะแก้ตารางเดิมเป็นแบบนี้:

| Text Feature          | คืออะไร                                     | ตัวอย่าง                     | ใช้ตอบคำถามอะไรได้                                    |
| --------------------- | ------------------------------------------- | ---------------------------- | ----------------------------------------------------- |
| **Review Length**     | จำนวนคำหรือตัวอักษร                         | 120 words                    | รีวิวที่ยาวกว่าได้รับ helpful vote มากกว่าหรือไม่     |
| **Sentiment Score**   | คะแนนความเป็นบวก/ลบ                         | `-0.72`, `+0.85`             | ความรู้สึกลูกค้าสัมพันธ์กับ rating หรือไม่            |
| **Aspect Indicators** | ระบุว่าพูดถึงหัวข้อใด                       | `delivery=1`, `price=0`      | ลูกค้าบ่นเรื่องใดมากที่สุด                            |
| **Lexical Diversity** | ความหลากหลายของคำ                           | unique words / total words   | รีวิวที่เขียนละเอียดมีประโยชน์มากกว่าหรือไม่          |
| **TF-IDF Vector**     | เวกเตอร์ที่ให้น้ำหนักคำเด่นในเอกสาร         | `[0, .35, 0, .81, ...]`      | คำใดแยกกลุ่มรีวิวบวกกับลบ                             |
| **Text Embedding**    | dense vector ที่แทนความหมายโดยรวมของข้อความ | `[.12, -.44, .08, ...]`      | รีวิวใดมีความหมายคล้ายกัน หรือมี customer themes อะไร |
| **Custom Feature**    | feature ที่ทีมสร้างตามโจทย์                 | `refund=1`, `defect_count=3` | ตอบ business question เฉพาะของทีม                     |

````python
# ===== ตัวอย่างที่ 1: Hacker News API (JSON, ไม่ต้องใช้ key) =====
# ได้หัวข้อข่าวพร้อม score และจำนวนคอมเมนต์ เหมาะกับการดูว่าเรื่องแบบไหนคนสนใจ
import urllib.request, json, time
import pandas as pd

def hn_get(url, timeout=30):
    with urllib.request.urlopen(url, timeout=timeout) as r:
        return json.loads(r.read().decode("utf-8"))

def fetch_hackernews(n=200, sleep=0.1):
    base = "https://hacker-news.firebaseio.com/v0"
    # ตัวอย่างนี้เก็บสูงสุด 200 stories ก็เพียงพอสำหรับ Part 3 เวอร์ชันลด workload
    ids = hn_get(f"{base}/topstories.json")[:n]
    rows = []
    for i, item_id in enumerate(ids):
        try:
            it = hn_get(f"{base}/item/{item_id}.json")
        except Exception as e:
            print("ข้าม", item_id, e)
            continue
        if not it or it.get("type") != "story" or not it.get("title"):
            continue
        rows.append({
            "doc_id": str(it["id"]),
            "text": it["title"] + (" " + it.get("text", "") if it.get("text") else ""),
            "score": it.get("score", 0),
            "n_comments": it.get("descendants", 0),
            "author": it.get("by"),
            "published_at": pd.to_datetime(it.get("time", 0), unit="s").isoformat(),
            "url": f"https://news.ycombinator.com/item?id={it['id']}",
            "source": "hackernews",
        })
        time.sleep(sleep)
        if (i + 1) % 100 == 0:
            print(f"ดึงแล้ว {i+1} รายการ")
    return rows

# rows = fetch_hackernews(50)
# print(len(rows)); rows[0]
````
