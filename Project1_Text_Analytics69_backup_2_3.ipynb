## เทคนิคที่อาจใช้ — เลือกเท่าที่จำเป็น

ไม่ต้องใช้ทั้งหมด โดยทั่วไป **2–4 เทคนิคก็เพียงพอ** ให้เลือกตามคำถามของทีม

| เทคนิค | เหมาะเมื่ออยากรู้ |
|---|---|
| **Word / n-gram frequency** | คนพูดถึงอะไรบ่อย |
| **TF-IDF / distinctive words** | สองกลุ่มพูดต่างกันอย่างไร |
| **Sentiment** | ข้อความเป็นบวก/ลบเพียงใด |
| **Aspect analysis** | คนพูดถึงราคา คุณภาพ การส่งของ ฯลฯ อย่างไร |
| **Topic modeling** | มีประเด็นอะไรที่เราอาจยังไม่รู้ |
| **Trend / segmentation** | ปัญหาเปลี่ยนตามเวลา หรือหนักในกลุ่มไหน |

> เทคนิคขั้นสูง เช่น embedding, clustering, regression, classification หรือ anomaly detection ทำได้ แต่ **ไม่บังคับ** และเหมาะสำหรับ Bonus เมื่อใช้ถูกและอธิบายได้

ถ้ายังไม่เคยใช้เทคนิคใด ให้ทดลองกับข้อมูลชุดเล็กก่อน แล้วตรวจด้วยตาว่าผลสมเหตุสมผลหรือไม่

> **หลักฐาน:** ตัวเลขพร้อม n และชี้ว่าดูได้จากกราฟหรือตารางไหน \

> **So what:** สินค้า 12 ชิ้นนี้คิดเป็น 18% ของยอดรีวิวทั้งหมวด ถ้าแก้ปัญหาฝาขวดได้ คาดว่าคะแนนเฉลี่ยของหมวดจะขยับขึ้นราว 0.15 ดาว \

ไม่ต้องทำ EDA ทุกตัวแปร


````python
# ============================================================
# 2.3.1 โหลดข้อมูลและเตรียมตัวแปรสำหรับการวิเคราะห์ (EDA)
# โหลดข้อมูลจาก Part_2/data/data.csv เพียงเท่านั้นตามคำสั่ง
# ============================================================
import os
import ast
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# กำหนดสไตล์ของกราฟ
plt.style.use('seaborn-v0_8-whitegrid' if 'seaborn-v0_8-whitegrid' in plt.style.available else 'default')
plt.rcParams['font.family'] = 'sans-serif'
plt.rcParams['font.sans-serif'] = ['DejaVu Sans', 'Arial', 'Tahoma']

# โหลดข้อมูลจาก Part_2/data/data.csv เพียงเท่านั้น
DATA_PATH = r"W:\Learn_Program\project_bigdata\Part_2\data\data.csv"
if not os.path.exists(DATA_PATH):
    DATA_PATH = os.path.join("Part_2", "data", "data.csv")

df = pd.read_csv(DATA_PATH)

# นิยามรีวิวเชิงลบ (1-2 ดาว) ซึ่งเป็นกลุ่มสำคัญตาม Client Brief 1 (ทีม Product Management)
df["negative_review"] = df["rating"] <= 2
df["word_count"] = df["text"].fillna("").astype(str).apply(lambda x: len(x.split()))

# สกัดชื่อกลุ่มสินค้า (Product Category) จากคอลัมน์ categories
def extract_category(cat_str):
    try:
        val = ast.literal_eval(cat_str)
        if isinstance(val, list) and len(val) > 0:
            return val[2] if len(val) > 2 else val[-1]
    except:
        pass
    return "Other / Unknown"

df["product_category"] = df["categories"].apply(extract_category)

print(" โหลดข้อมูลสำเร็จจาก:", DATA_PATH)
print(f" จำนวนแถวทั้งหมด: {len(df):,} รายการ, จำนวนคอลัมน์: {df.shape[1]}")
print(f" จำนวนรีวิวเชิงลบ (1–2 ดาว): {df['negative_review'].sum():,} รายการ ({df['negative_review'].mean()*100:.2f}%)")
````

````python
# ============================================================
# 2.3.2 EDA 1: การแจกแจงดาวและสัดส่วนรีวิวเชิงลบ (Rating Distribution)
# ============================================================
# 1. ตารางสรุปการแจกแจงคะแนนรีวิว 1-5 ดาว
rating_dist = df["rating"].value_counts().sort_index().to_frame("จำนวนรีวิว")
rating_dist["ร้อยละ (%)"] = (rating_dist["จำนวนรีวิว"] / len(df) * 100).round(2)
rating_dist["ระดับความพึงพอใจ"] = [
    "รีวิวเชิงลบมาก (Critical Negative)",
    "รีวิวเชิงลบ (Negative)",
    "ปานกลาง (Neutral)",
    "ดี (Positive)",
    "ดีเยี่ยม (High Positive)"
]

# สรุปภาพรวมกลุ่มรีวิวเชิงลบ (1-2 ดาว) เทียบกับทั่วไป
neg_count = int(df["negative_review"].sum())
neg_pct = float(df["negative_review"].mean() * 100)
summary_overview = pd.DataFrame({
    "กลุ่มรีวิว": ["รีวิวทั้งหมด", "รีวิวเชิงลบ (1–2 ดาว)", "รีวิวทั่วไป/เชิงบวก (3–5 ดาว)"],
    "จำนวน": [len(df), neg_count, len(df) - neg_count],
    "ร้อยละ (%)": [100.0, round(neg_pct, 2), round(100 - neg_pct, 2)]
})

display(rating_dist)
display(summary_overview)

# 2. กราฟแท่งแสดงการกระจายตัวของคะแนนรีวิว
fig, ax = plt.subplots(figsize=(8, 4.8))
ratings = rating_dist.index.astype(int)
counts = rating_dist["จำนวนรีวิว"].values
colors = ['#d9534f', '#f0ad4e', '#5bc0de', '#428bca', '#5cb85c']

bars = ax.bar([f"{r} Stars" for r in ratings], counts, color=colors, width=0.6, edgecolor='#333333', linewidth=0.8)

for bar in bars:
    h = bar.get_height()
    pct = h / len(df) * 100
    ax.annotate(f"{h:,}\n({pct:.1f}%)",
                xy=(bar.get_x() + bar.get_width() / 2, h),
                xytext=(0, 4), textcoords="offset points",
                ha='center', va='bottom', fontsize=9.5, fontweight='bold')

ax.set_title("EDA 1: Distribution of Review Ratings (N = 30,000)", fontsize=13, fontweight='bold', pad=14)
ax.set_xlabel("Review Rating", fontsize=11)
ax.set_ylabel("Number of Reviews", fontsize=11)
ax.set_ylim(0, max(counts) * 1.18)
plt.tight_layout()
plt.show()
````

**ข้อสังเกตจาก EDA 1 (การแจกแจงดาว):**
* แม้คะแนนรีวิวโดยรวมจะกระจุกตัวที่ 5 ดาวมากที่สุด (67.53%) แต่พบว่ามี **รีวิวเชิงลบระดับ 1–2 ดาว รวมกันถึง 4,660 รายการ หรือคิดเป็น 15.53%** (เฉลี่ยประมาณ 1 ใน 6 รีวิว)
* สำหรับทีม Product Management ลูกค้ากลุ่มนี้คือ Voice of Customer ที่มีความสำคัญสูงสุด เพราะสะท้อนถึงลูกค้าที่ไม่พึงพอใจและกำลังประสบปัญหากับสินค้าจริง การแก้ไขปัญหากลุ่มนี้จึงมีโอกาสยกระดับคะแนนภาพรวมของสินค้าได้มากที่สุด


````python
# ============================================================
# 2.3.3 EDA 2: สัดส่วนรีวิวเชิงลบแยกตามกลุ่มสินค้า (Category vs. Negative Rate)
# ============================================================
# คำนวณสัดส่วนรีวิวเชิงลบแยกตามหมวดหมู่สินค้า
cat_summary = df.groupby("product_category").agg(
    total_reviews=("rating", "count"),
    neg_reviews=("negative_review", "sum"),
    neg_rate=("negative_review", "mean"),
    avg_rating=("rating", "mean")
).reset_index()

cat_summary["neg_rate_pct"] = (cat_summary["neg_rate"] * 100).round(2)
cat_summary["avg_rating"] = cat_summary["avg_rating"].round(2)

# กรองเฉพาะหมวดที่มีรีวิว >= 100 รายการ เพื่อความน่าเชื่อถือทางสถิติ
cat_filtered = cat_summary[cat_summary["total_reviews"] >= 100].sort_values(by="neg_rate_pct", ascending=False)

# แสดง 10 อันดับแรก
display_cat = cat_filtered.head(10).rename(columns={
    "product_category": "หมวดหมู่สินค้า (Product Category)",
    "total_reviews": "จำนวนรีวิวทั้งหมด",
    "neg_reviews": "รีวิวเชิงลบ (1–2 ดาว)",
    "neg_rate_pct": "สัดส่วนเชิงลบ (%)",
    "avg_rating": "คะแนนเฉลี่ย"
})[["หมวดหมู่สินค้า (Product Category)", "จำนวนรีวิวทั้งหมด", "รีวิวเชิงลบ (1–2 ดาว)", "สัดส่วนเชิงลบ (%)", "คะแนนเฉลี่ย"]]
display(display_cat.reset_index(drop=True))

# วาดกราฟแท่งแนวนอน Top 10 Product Categories by Negative Review Rate
top10_cat = cat_filtered.head(10).sort_values(by="neg_rate_pct", ascending=True)

fig, ax = plt.subplots(figsize=(10, 5.5))
bar_colors = ['#c9302c' if rate >= 20 else '#ec971f' for rate in top10_cat["neg_rate_pct"]]
bars = ax.barh(top10_cat["product_category"], top10_cat["neg_rate_pct"], color=bar_colors, height=0.65, edgecolor='#333333', linewidth=0.7)

# เส้น Benchmark ค่าเฉลี่ยรวม
benchmark_rate = df["negative_review"].mean() * 100
ax.axvline(benchmark_rate, color='#333333', linestyle='--', linewidth=1.5, label=f"Office Products Overall Avg ({benchmark_rate:.1f}%)")

for bar in bars:
    w = bar.get_width()
    ax.annotate(f"{w:.1f}%",
                xy=(w, bar.get_y() + bar.get_height() / 2),
                xytext=(6, 0), textcoords="offset points",
                ha='left', va='center', fontsize=9.5, fontweight='bold')

ax.set_title("EDA 2: Top Product Categories by Negative Review Rate (Min 100 Reviews)", fontsize=13, fontweight='bold', pad=14)
ax.set_xlabel("Negative Review Rate (%) [1-2 Stars]", fontsize=11)
ax.set_ylabel("Product Category", fontsize=11)
ax.set_xlim(0, max(top10_cat["neg_rate_pct"]) * 1.18)
ax.legend(loc="lower right", fontsize=10)
plt.tight_layout()
plt.show()
````

* ผลวิเคราะห์ชี้ชัดว่าสินค้ากลุ่ม **อุปกรณ์และเฟอร์นิเจอร์สำนักงาน** มีอัตราการเกิดรีวิวเชิงลบสูงกว่าเกณฑ์เฉลี่ยรวม (15.5%) อย่างมีนัยสำคัญ:

````python
# ============================================================
# 2.3.4 EDA 3: พฤติกรรมการเขียนรีวิว: ความยาวรีวิวและ Helpful Votes แยกตาม Rating
# ============================================================
# วิเคราะห์ความยาวรีวิว (Word Count) และคะแนน Helpful Vote แยกตาม Rating
text_eda = df.groupby("rating").agg(
    total_reviews=("rating", "count"),
    avg_word_count=("word_count", "mean"),
    median_word_count=("word_count", "median"),
    avg_helpful_vote=("helpful_vote", "mean"),
    median_helpful_vote=("helpful_vote", "median")
).round(2).reset_index()

display_text_eda = text_eda.rename(columns={
    "rating": "คะแนน (Rating)",
    "total_reviews": "จำนวนรีวิว",
    "avg_word_count": "ความยาวเฉลี่ย (คำ)",
    "median_word_count": "ความยาวมัธยฐาน (คำ)",
    "avg_helpful_vote": "Helpful Vote เฉลี่ย",
    "median_helpful_vote": "Helpful Vote มัธยฐาน"
})
display(display_text_eda)

# กราฟเปรียบเทียบความยาวรีวิวและ Helpful Vote
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 4.8))
bar_colors = ['#d9534f', '#f0ad4e', '#5bc0de', '#428bca', '#5cb85c']

# Subplot 1: Median Word Count
b1 = ax1.bar([f"{int(r)} Stars" for r in text_eda["rating"]], text_eda["median_word_count"], 
             color=bar_colors, width=0.55, edgecolor='#333333', linewidth=0.7)
for b in b1:
    h = b.get_height()
    ax1.annotate(f"{int(h)} words", xy=(b.get_x() + b.get_width() / 2, h),
                 xytext=(0, 4), textcoords="offset points", ha='center', va='bottom', fontsize=9.5, fontweight='bold')
ax1.set_title("Median Review Length (Words) by Rating", fontsize=12, fontweight='bold', pad=12)
ax1.set_xlabel("Rating", fontsize=10)
ax1.set_ylabel("Median Word Count", fontsize=10)
ax1.set_ylim(0, max(text_eda["median_word_count"]) * 1.22)

# Subplot 2: Average Helpful Vote
b2 = ax2.bar([f"{int(r)} Stars" for r in text_eda["rating"]], text_eda["avg_helpful_vote"], 
             color=bar_colors, width=0.55, edgecolor='#333333', linewidth=0.7)
for b in b2:
    h = b.get_height()
    ax2.annotate(f"{h:.2f}", xy=(b.get_x() + b.get_width() / 2, h),
                 xytext=(0, 4), textcoords="offset points", ha='center', va='bottom', fontsize=9.5, fontweight='bold')
ax2.set_title("Average Helpful Votes by Rating", fontsize=12, fontweight='bold', pad=12)
ax2.set_xlabel("Rating", fontsize=10)
ax2.set_ylabel("Average Helpful Votes", fontsize=10)
ax2.set_ylim(0, max(text_eda["avg_helpful_vote"]) * 1.25)

plt.suptitle("EDA 3: Review Depth & Social Influence by Rating", fontsize=13, fontweight='bold', y=1.02)
plt.tight_layout()
plt.show()
````

### 2.4 เตรียมข้อความและสร้าง Text Features

ก่อนนำข้อความไปวิเคราะห์ จำเป็นต้องมีการ **preprocessing** เพื่อทำให้ข้อความอยู่ในรูปที่เหมาะสมกับเทคนิคที่จะใช้ อย่างไรก็ตาม ในงานนี้ไม่จำเป็นต้องทำ preprocessing ทุกขั้นตอน ให้เลือกทำ **เท่าที่จำเป็น** ตามวัตถุประสงค์ของการวิเคราะห์

ตัวอย่างเช่น การเปลี่ยนข้อความเป็นตัวพิมพ์เล็ก (lower case) การนำ URL หรือ HTML ออก การแบ่งข้อความเป็นคำ (tokenization) หรือการตัด stopwords เช่น `the`, `and`, `is` ออก ในกรณีที่คำเหล่านี้ไม่ได้ช่วยในการวิเคราะห์

**ไม่บังคับให้ใช้ stemming หรือ lemmatization** หากไม่มีเหตุผลว่าขั้นตอนดังกล่าวจะช่วยตอบคำถามของทีมได้ดีขึ้น

หลังจากเตรียมข้อความแล้ว ให้สร้าง **Text Features อย่างน้อย 2 ชนิด** เพื่อนำข้อความมาเปลี่ยนให้อยู่ในรูปข้อมูลที่สามารถนำไปสำรวจ เปรียบเทียบ หรือวิเคราะห์ต่อได้

---

#### Text Feature คืออะไร?

Text Feature คือ **ข้อมูลเชิงตัวเลขที่สกัดหรือสร้างขึ้นจากข้อความ** เพื่อแทนลักษณะหรือความหมายบางอย่างของข้อความนั้น

Text Feature อาจอยู่ในรูปของ

* **ค่าตัวเลขค่าเดียว** เช่น จำนวนคำ หรือ sentiment score
* **ตัวแปรหลายตัว** เช่น `price=1`, `quality=0`, `delivery=1`
* **เวกเตอร์หลายมิติ** เช่น TF-IDF vector หรือ Text Embedding

ตัวอย่างเช่น รีวิว

> `"The product is easy to use but the delivery was very slow."`

สามารถเปลี่ยนเป็น feature ได้หลายรูปแบบ

| Feature                    |                 ตัวอย่างค่าที่ได้ |
| -------------------------- | --------------------------------: |
| จำนวนคำในรีวิว             |                                12 |
| Sentiment score            |                             -0.15 |
| กล่าวถึง usability หรือไม่ |                                 1 |
| กล่าวถึง delivery หรือไม่  |                                 1 |
| กล่าวถึง price หรือไม่     |                                 0 |
| TF-IDF vector              |   `[0.00, 0.41, 0.00, 0.72, ...]` |
| Embedding vector           | `[0.21, -0.08, 0.63, ..., -0.14]` |

ดังนั้นข้อความหนึ่งข้อความสามารถถูกแทนได้ทั้งด้วย **feature ที่ตีความได้โดยตรง** และ **feature แบบเวกเตอร์ที่ใช้แทนเนื้อหาหรือความหมายของข้อความ**

---

## ตัวอย่าง Text Features ที่สามารถเลือกใช้

ผมจะแก้ตารางเดิมเป็นแบบนี้:

| Text Feature          | คืออะไร                                     | ตัวอย่าง                     | ใช้ตอบคำถามอะไรได้                                    |
| --------------------- | ------------------------------------------- | ---------------------------- | ----------------------------------------------------- |
| **Review Length**     | จำนวนคำหรือตัวอักษร                         | 120 words                    | รีวิวที่ยาวกว่าได้รับ helpful vote มากกว่าหรือไม่     |
| **Sentiment Score**   | คะแนนความเป็นบวก/ลบ                         | `-0.72`, `+0.85`             | ความรู้สึกลูกค้าสัมพันธ์กับ rating หรือไม่            |
| **Aspect Indicators** | ระบุว่าพูดถึงหัวข้อใด                       | `delivery=1`, `price=0`      | ลูกค้าบ่นเรื่องใดมากที่สุด                            |
| **Lexical Diversity** | ความหลากหลายของคำ                           | unique words / total words   | รีวิวที่เขียนละเอียดมีประโยชน์มากกว่าหรือไม่          |
| **TF-IDF Vector**     | เวกเตอร์ที่ให้น้ำหนักคำเด่นในเอกสาร         | `[0, .35, 0, .81, ...]`      | คำใดแยกกลุ่มรีวิวบวกกับลบ                             |
| **Text Embedding**    | dense vector ที่แทนความหมายโดยรวมของข้อความ | `[.12, -.44, .08, ...]`      | รีวิวใดมีความหมายคล้ายกัน หรือมี customer themes อะไร |
| **Custom Feature**    | feature ที่ทีมสร้างตามโจทย์                 | `refund=1`, `defect_count=3` | ตอบ business question เฉพาะของทีม                     |

````python
# ============================================================
# 2.4.1 Text Preprocessing
# ============================================================
import re
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

def clean_text(text):
    """
    ฟังก์ชันทำความสะอาดข้อความรีวิวเท่าที่จำเป็นสำหรับการวิเคราะห์:
    - จัดการค่า missing (NaN)
    - แปลงตัวอักษรเป็นพิมพ์เล็ก (Lowercasing)
    - ลบ HTML tags และ URLs
    - จัดการช่องว่างซ้ำซ้อน
    """
    if pd.isna(text):
        return ""
    text = str(text).lower()
    text = re.sub(r"<[^>]+>", " ", text)
    text = re.sub(r"https?://\S+|www\.\S+", " ", text)
    text = re.sub(r"\s+", " ", text).strip()
    return text

# นำไปประมวลผลกับคอลัมน์ text
df["clean_text"] = df["text"].apply(clean_text)

print(" ทำความสะอาดข้อความรีวิวสำเร็จ!")
print(f" จำนวนข้อความทั้งหมด: {len(df):,} รายการ")
print(f" จำนวนข้อความที่เป็นค่าว่างหลังทำความสะอาด: {(df['clean_text'] == '').sum():,} รายการ")

# แสดงตัวอย่างการเปรียบเทียบก่อนและหลังทำความสะอาด
sample_comp = df[["text", "clean_text"]].dropna().head(5).copy()
sample_comp.columns = ["Original Text (ข้อความเดิม)", "Cleaned Text (ข้อความที่ทำความสะอาดแล้ว)"]
display(sample_comp)
````

````python
# ============================================================
# 2.4.2 Text Feature 1: Sentiment Score (VADER Lexicon Analyzer)
# คำถามธุรกิจ: ความรู้สึกลูกค้าสัมพันธ์กับ rating หรือไม่
# ============================================================
import sys
import os
# เพิ่ม path ของ venv เพื่อนำเข้า nltk
venv_site = r"W:\Learn_Program\project_bigdata\.venv\Lib\site-packages"
if os.path.exists(venv_site) and venv_site not in sys.path:
    sys.path.insert(0, venv_site)

import nltk
from nltk.sentiment import SentimentIntensityAnalyzer

# ดาวน์โหลด VADER lexicon (หากยังไม่มี)
nltk.download("vader_lexicon", quiet=True)
sia = SentimentIntensityAnalyzer()

# 1. คำนวณ Sentiment Score (Compound Score ระหว่าง -1.0 ถึง +1.0)
df["sentiment_score"] = df["clean_text"].apply(lambda x: sia.polarity_scores(x)["compound"])

# 2. จำแนกกลุ่มความรู้สึกตามเกณฑ์มาตรฐาน VADER
def sentiment_category(score):
    if score >= 0.05:
        return "Positive"
    elif score <= -0.05:
        return "Negative"
    else:
        return "Neutral"

df["sentiment_label"] = df["sentiment_score"].apply(sentiment_category)

# 3. วิเคราะห์ความสัมพันธ์ระหว่าง Rating กับ Sentiment Score
corr = df["rating"].corr(df["sentiment_score"])

sentiment_by_rating = df.groupby("rating").agg(
    จำนวนรีวิว=("rating", "count"),
    Sentiment_เฉลี่ย=("sentiment_score", "mean"),
    Sentiment_มัธยฐาน=("sentiment_score", "median"),
    สัดส่วนเชิงลบ_pct=("sentiment_label", lambda x: (x == "Negative").mean() * 100),
    สัดส่วนเชิงบวก_pct=("sentiment_label", lambda x: (x == "Positive").mean() * 100)
).round(3).reset_index()

sentiment_by_rating.columns = [
    "คะแนน (Rating)", "จำนวนรีวิว", "Sentiment เฉลี่ย", "Sentiment มัธยฐาน", 
    "สัดส่วน Negative (%)", "สัดส่วน Positive (%)"
]

print(f" ค่าสหสัมพันธ์ (Pearson Correlation) ระหว่าง Rating และ Sentiment Score: {corr:.4f}")
display(sentiment_by_rating)

# 4. กราฟเปรียบเทียบ Sentiment Score ตามระดับ Rating
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 4.8))

colors = ['#d9534f', '#f0ad4e', '#5bc0de', '#428bca', '#5cb85c']
ratings = sentiment_by_rating["คะแนน (Rating)"].astype(int)

# กราฟ 1: ค่าเฉลี่ย Sentiment Score
bars1 = ax1.bar([f"{r} Stars" for r in ratings], sentiment_by_rating["Sentiment เฉลี่ย"], color=colors, width=0.55, edgecolor='#333333', linewidth=0.7)
ax1.axhline(0, color='gray', linestyle='--', linewidth=1)
for b in bars1:
    h = b.get_height()
    va = 'bottom' if h >= 0 else 'top'
    ypos = h + 0.03 if h >= 0 else h - 0.06
    ax1.annotate(f"{h:+.2f}", xy=(b.get_x() + b.get_width() / 2, ypos),
                 ha='center', va=va, fontsize=9.5, fontweight='bold')
ax1.set_title("Average Sentiment Score by Rating", fontsize=12, fontweight='bold', pad=12)
ax1.set_xlabel("Review Rating", fontsize=10)
ax1.set_ylabel("Compound Sentiment Score (-1 to +1)", fontsize=10)
ax1.set_ylim(-0.35, 0.8)

# กราฟ 2: สัดส่วน Negative vs Positive (%)
b_pos = ax2.bar([f"{r} Stars" for r in ratings], sentiment_by_rating["สัดส่วน Positive (%)"], 
                label="Positive Sentiment (Score >= 0.05)", color='#428bca', width=0.55, edgecolor='#333333', linewidth=0.7)
b_neg = ax2.bar([f"{r} Stars" for r in ratings], sentiment_by_rating["สัดส่วน Negative (%)"], 
                bottom=sentiment_by_rating["สัดส่วน Positive (%)"],
                label="Negative Sentiment (Score <= -0.05)", color='#d9534f', width=0.55, edgecolor='#333333', linewidth=0.7)
ax2.set_title("Sentiment Composition by Rating", fontsize=12, fontweight='bold', pad=12)
ax2.set_xlabel("Review Rating", fontsize=10)
ax2.set_ylabel("Percentage (%)", fontsize=10)
ax2.set_ylim(0, 115)
ax2.legend(loc="upper left", fontsize=9)

plt.suptitle(f"Feature 1: Customer Sentiment vs Rating (Correlation r = {corr:.2f})", fontsize=13, fontweight='bold', y=1.02)
plt.tight_layout()
plt.show()
````

**ข้อสังเกตจาก Feature 1 (Sentiment Score):**
* **ความสัมพันธ์ระหว่างความรู้สึกกับคะแนนดาว:** ค่า Correlation ระหว่าง Rating และ Sentiment Score อยู่ที่ **+0.5476** แสดงถึงความสัมพันธ์เชิงบวกระดับปานกลางค่อนข้างสูง (Moderate-to-Strong Positive Association)
* **ความสอดคล้องของรีวิว 5 ดาว:** รีวิว 5 ดาวมี Sentiment เฉลี่ยสูงถึง **+0.598** (โดยเป็น Positive สูงถึง 86.74% และ Negative เพียง 3.38%)
* **ความเข้มข้นของรีวิว 1 ดาว:** รีวิว 1 ดาวมีค่า Sentiment เฉลี่ยติดลบที่ **-0.150 (มัธยฐาน -0.178)** และมีสัดส่วนรีวิวอารมณ์เชิงลบสูงถึง **54.16%**
* **ตอบคำถามธุรกิจ:** ความรู้สึกลูกค้าสัมพันธ์กับ Rating อย่างมีนัยสำคัญ แต่ Sentiment Score จากข้อความช่วยเปิดเผยมิติอารมณ์ที่ลึกกว่าตัวเลขดาว เช่น กรณีรีวิวที่ให้ดาวปานกลาง (3 ดาว) แต่มีเนื้อความบ่นเรื่องปัญหาจริงจัง ซึ่งตัวเลขคะแนนดาวอย่างเดียวไม่สามารถสะท้อนได้


````python
# ============================================================
# 2.4.3 Text Feature 2: Aspect Indicators (Binary Topic Features)
# คำถามธุรกิจ: ลูกค้าบ่นเรื่องใดมากที่สุด
# ============================================================
# กำหนด Dictionary คำศัพท์สำหรับสกัดประเด็น (Aspects) ที่เกี่ยวข้องกับ Brief 1
aspects_dict = {
    # 1. ด้านการทำงาน/ชิ้นส่วนกลไก/ฟังก์ชัน (Functionality & Operations)
    "functionality": [
        "works", "work", "working", "print", "printing", "jam", "jammed", 
        "ink", "cartridge", "connect", "connection", "setup", "software", "error"
    ],
    # 2. ด้านคุณภาพและความคงทนของวัสดุ (Product Quality & Durability)
    "quality": [
        "quality", "durable", "durability", "sturdy", "flimsy", "broke", 
        "broken", "cheap", "crack", "cracked", "defect", "defective", "poor quality"
    ],
    # 3. ด้านราคาและความคุ้มค่า (Price & Value)
    "price": [
        "price", "expensive", "cost", "value", "worth", "money", "affordable"
    ],
    # 4. ด้านการบริการลูกค้าและการเคลม/คืนสินค้า (Customer Service & Returns)
    "service": [
        "service", "support", "refund", "return", "returned", "warranty", "replace", "replacement"
    ],
    # 5. ด้านการจัดส่งและบรรจุภัณฑ์ (Delivery & Packaging)
    "delivery": [
        "delivery", "shipping", "shipped", "arrive", "arrived", "package", "packaging", "box"
    ]
}

# สร้าง Binary Features (0 หรือ 1) สำหรับแต่ละ Aspect
aspect_cols = []
for aspect, keywords in aspects_dict.items():
    col_name = f"aspect_{aspect}"
    aspect_cols.append(col_name)
    pattern = r"\b(?:" + "|".join(re.escape(k) for k in keywords) + r")\b"
    df[col_name] = df["clean_text"].str.contains(pattern, regex=True, na=False).astype(int)

print(" สร้าง Aspect Indicators สำเร็จ 5 ประเด็น:")
for col in aspect_cols:
    print(f"  - {col}: มีรีวิวพูดถึง {df[col].sum():,} รายการ ({df[col].mean()*100:.2f}%)")

# 1. ตารางเปรียบเทียบสัดส่วนการกล่าวถึงใน รีวิวทั้งหมด vs รีวิวเชิงลบ (1-2 ดาว)
df_neg = df[df["rating"] <= 2]

aspect_comparison = pd.DataFrame({
    "Aspect (ประเด็น)": [
        "Functionality (การทำงาน/ฟังก์ชัน)",
        "Quality (คุณภาพ/ความคงทน)",
        "Price & Value (ราคา/ความคุ้มค่า)",
        "Service & Warranty (บริการ/การคืนของ)",
        "Delivery & Packaging (การจัดส่ง/พัสดุ)"
    ],
    "Mention Rate รวม (%)": (df[aspect_cols].mean() * 100).round(2).values,
    "Mention Rate รีวิวเชิงลบ (%)": (df_neg[aspect_cols].mean() * 100).round(2).values,
    "คะแนนเฉลี่ยเมื่อพูดถึง": [df[df[c] == 1]["rating"].mean() for c in aspect_cols],
    "Negative Rate (%)": [df[df[c] == 1]["negative_review"].mean() * 100 for c in aspect_cols]
}).round(2).sort_values(by="Mention Rate รีวิวเชิงลบ (%)", ascending=False)

display(aspect_comparison.reset_index(drop=True))

# 2. กราฟเปรียบเทียบ Aspect Mention Rate
fig, ax = plt.subplots(figsize=(10, 5.2))
y_positions = np.arange(len(aspect_comparison))
bar_width = 0.38

b1 = ax.barh(y_positions + bar_width/2, aspect_comparison["Mention Rate รีวิวเชิงลบ (%)"], 
             height=bar_width, color='#d9534f', edgecolor='#333333', linewidth=0.7, label="Negative Reviews (1-2 Stars, n=4,660)")
b2 = ax.barh(y_positions - bar_width/2, aspect_comparison["Mention Rate รวม (%)"], 
             height=bar_width, color='#428bca', edgecolor='#333333', linewidth=0.7, label="All Reviews (N=30,000)")

for b in b1:
    w = b.get_width()
    ax.annotate(f"{w:.1f}%", xy=(w, b.get_y() + b.get_height() / 2),
                xytext=(5, 0), textcoords="offset points", ha='left', va='center', fontsize=9, fontweight='bold', color='#c9302c')

for b in b2:
    w = b.get_width()
    ax.annotate(f"{w:.1f}%", xy=(w, b.get_y() + b.get_height() / 2),
                xytext=(5, 0), textcoords="offset points", ha='left', va='center', fontsize=8.5, color='#2c3e50')

ax.set_yticks(y_positions)
ax.set_yticklabels(aspect_comparison["Aspect (ประเด็น)"], fontsize=10)
ax.invert_yaxis()
ax.set_xlabel("Aspect Mention Rate (%)", fontsize=11)
ax.set_xlim(0, max(aspect_comparison["Mention Rate รีวิวเชิงลบ (%)"]) * 1.18)
ax.set_title("Feature 2: Customer Issues Breakdown (All vs. Negative Reviews)", fontsize=13, fontweight='bold', pad=14)
ax.legend(loc="lower right", fontsize=10)
plt.tight_layout()
plt.show()
````

**ข้อสังเกตจาก Feature 2 (Aspect Indicators):**
* **ตอบคำถามธุรกิจ: ลูกค้าบ่นเรื่องใดมากที่สุด**
  1. **อันดับ 1 — ปัญหาด้านการทำงานและฟังก์ชันสินค้า (Functionality):** ในกลุ่มรีวิวเชิงลบ (1–2 ดาว) มีสัดส่วนการบ่นเรื่องนี้สูงที่สุดถึง **33.22%** (ลูกค้าบ่นเรื่องเครื่องพิมพ์ไม่ออก, กระดาษติด, หมึกไม่ออก, เชื่อมต่อไม่ได้ หรือซอฟต์แวร์ error)
  2. **อันดับ 2 — ปัญหาด้านคุณภาพและความคงทนของวัสดุ (Quality):** ถูกกล่าวถึงในรีวิวเชิงลบสูงถึง **21.31%** (สินค้าพังง่าย, แตกหัก, ชิ้นส่วนก๊องแก๊ง ไม่ทนทาน)
  3. **อันดับ 3 — ราคาและความคุ้มค่า (Price & Value):** คิดเป็น **19.74%** ของรีวิวเชิงลบ (รู้สึกราคาแพงเกินจริง ไม่คุ้มเงินที่จ่าย)
  4. **อันดับ 4 — การบริการลูกค้าและการเคลมคืนสินค้า (Service & Support):** คิดเป็น **17.58%** แต่ที่น่าตกใจคือ *รีวิวที่พูดถึงเรื่อง Service ได้คะแนนเฉลี่ยต่ำที่สุดในบรรดาทุก Aspect เพียง 3.05 ดาว และมีอัตราส่วนรีวิวเชิงลบสูงถึง 45.5%* สะท้อนว่าเมื่อสินค้ามีปัญหาแล้วลูกค้าขอคืนหรือเคลมยาก จะทำให้ความไม่พอใจทวีความรุนแรง
  5. **อันดับ 5 — การจัดส่งและพัสดุ (Delivery & Packaging):** มีเพียง **10.19%** ในรีวิวเชิงลบ
* **คุณค่าต่อ Client Brief 1:** ชี้ชัดเจนว่าทีมพัฒนาผลิตภัณฑ์ (Product Team) ต้องนำงบประมาณ R&D ไปโฟกัสแก้ **ปัญหาการทำงานของสินค้า (Functionality)** และ **ความคงทนของวัสดุ (Quality)** เป็น 2 เรื่องหลัก เพราะเป็นสาเหตุที่ทำให้ลูกค้าไม่พอใจมากที่สุด

### **รายงานสรุปหัวข้อ 2.4: การเตรียมข้อความและสร้าง Text Features**

#### **1. ทำ Preprocessing อะไร และทำไม**
* **เทคนิคที่เลือกใช้:**
  1. **Lowercasing (แปลงข้อความเป็นตัวพิมพ์เล็ก):** เพื่อลดความซ้ำซ้อนของคำ ทำให้คำเดียวกันที่มีตัวพิมพ์ใหญ่-เล็กต่างกัน (เช่น `Price`, `price`, `PRICE` หรือ `Delivery`, `delivery`) ถูกมองเป็นคำเดียวกัน ช่วยให้การ match คำใน Aspect Indicators ทำได้อย่างสมบูรณ์
  2. **HTML & URL Removal (ตัดแท็ก HTML และลิงก์):** กำจัด Noise เช่น `<br>`, `http://...` ที่ไม่มีข้อมูลความรู้สึก
  3. **Whitespace Normalization (จัดระเบียบช่องว่าง):** ตัดช่องว่างซ้ำซ้อนเพื่อให้ข้อความสะอาดพร้อมต่อการวิเคราะห์
* **เทคนิคที่ไม่ใช้และเหตุผล:**
  1. **ไม่ใช้ Stemming / Lemmatization:** เนื่องจากเครื่องมือ VADER ทำงานได้ดีที่สุดกับคำเต็มที่มีการสะกดตามธรรมชาติ การตัดรากคำจะทำลายคุณลักษณะทางภาษาและทำให้ Sentiment Score คลาดเคลื่อน

#### Step 1 — กำหนดเว็บไซต์และสิ่งที่ต้องการเก็บ

เริ่มจากกำหนดให้ชัดก่อนว่าเราจะเก็บอะไร

ในตัวอย่างนี้
* `LISTING_URL` = หน้ารวมข่าว ใช้สำหรับ **หา link**
* `ARTICLE_PATH` = รูปแบบ URL ที่ถือว่าเป็น **บทความจริง**
* `MAX_ARTICLES` = จำนวนข่าวที่ทดลองเก็บ
* `DELAY` = เวลาพักระหว่างการเรียกแต่ละหน้า เพื่อไม่ยิง request ถี่เกินไป

จุดสำคัญคือ **หน้ารวมข่าวไม่ใช่ record ที่จะเอาไปวิเคราะห์** แต่เป็นเพียงทางเข้าไปหา article URLs


````python
# ===== Step 1: ตั้งค่า =====
# หากยังไม่มี package:
# !pip install -q requests beautifulsoup4

import re
import time
import requests
import pandas as pd

from bs4 import BeautifulSoup
from urllib.parse import urljoin, urlparse, urldefrag
from urllib.robotparser import RobotFileParser

BASE_URL = "https://www.thaipbs.or.th"
LISTING_URL = "https://www.thaipbs.or.th/news"

# Thai PBS ใช้ URL ข่าวลักษณะ /news/content/509281
ARTICLE_PATH_PATTERN = r"^/news/content/\d+$"

MAX_ARTICLES = 20     # เริ่มน้อย ๆ เพื่อทดสอบก่อน
DELAY = 1.5           # พัก 1.5 วินาทีระหว่างหน้า
BOT_USER_AGENT = "KKU-TextAnalytics-StudentBot/1.0"

session = requests.Session()
session.headers.update({"User-Agent": BOT_USER_AGENT})

print("Listing page:", LISTING_URL)
print("เป้าหมาย: 1 row = 1 news article")
````

#### Step 2 — ทำความสะอาด URL และตรวจ `robots.txt`

URL เดียวกันอาจเขียนได้หลายแบบ เช่น

`https://www.thaipbs.or.th/news`  
`https://www.thaipbs.or.th/news#main-content`

ส่วนหลัง `#...` เรียกว่า **fragment** และโดยทั่วไปไม่ได้หมายถึงหน้าใหม่ จึงควรตัดออกก่อนตรวจ URL ซ้ำ

จากนั้นตรวจ `robots.txt` ก่อนเข้าแต่ละส่วนของเว็บ ถ้าไม่อนุญาตให้ bot เข้า path นั้น ให้ **เปลี่ยนแหล่งข้อมูล** ไม่ใช่หาวิธีหลบ


````python
# ===== Step 2: normalize URL + robots.txt =====

_robots_cache = {}

def normalize_url(url):
    """ตัด #fragment ออกจาก URL เพื่อไม่ให้หน้าเดียวกันถูกนับซ้ำ"""
    clean_url, _ = urldefrag(url)
    return clean_url.rstrip("/")


def robots_allowed(url, user_agent=BOT_USER_AGENT):
    """คืน True เมื่อ robots.txt อนุญาตให้ bot เข้าถึง URL"""
    p = urlparse(url)
    root = f"{p.scheme}://{p.netloc}"

    if root not in _robots_cache:
        rp = RobotFileParser()
        rp.set_url(urljoin(root, "/robots.txt"))
        try:
            rp.read()
            _robots_cache[root] = rp
        except Exception as e:
            print("อ่าน robots.txt ไม่ได้:", e)
            return False

    return _robots_cache[root].can_fetch(user_agent, url)


print("robots.txt อนุญาตหน้ารวมข่าวหรือไม่:", robots_allowed(LISTING_URL))
````

#### Step 3 — เข้า “หน้ารวมข่าว” เพื่อหาเฉพาะ Article URLs

ขั้นนี้ **ยังไม่เก็บเนื้อหาข่าว**

Robot เปิดหน้ารวมข่าว แล้วดู `<a href="...">` ทุก link จากนั้นกรองให้เหลือเฉพาะ URL ที่ตรงกับรูปแบบบทความ เช่น

`/news/content/509281`

การกรองนี้คือสิ่งที่ทำให้ Robot ต่างจาก generic crawler ที่วิ่งไปทุกหน้าใน domain


````python
# ===== Step 3: หา link ของ "บทความจริง" =====

def find_article_links(listing_url):
    if not robots_allowed(listing_url):
        raise PermissionError("robots.txt ไม่อนุญาตให้เข้าหน้ารวมข่าวนี้")

    r = session.get(listing_url, timeout=20)
    r.raise_for_status()

    soup = BeautifulSoup(r.text, "html.parser")
    article_urls = set()

    for a in soup.find_all("a", href=True):
        url = urljoin(listing_url, a["href"])
        url = normalize_url(url)

        p = urlparse(url)

        # 1) เอาเฉพาะ domain เดียวกับเว็บไซต์หลัก
        same_domain = (p.netloc == urlparse(BASE_URL).netloc)

        # 2) เอาเฉพาะ path ที่มีรูปแบบเป็นบทความข่าว
        is_article = bool(re.match(ARTICLE_PATH_PATTERN, p.path))

        if same_domain and is_article:
            article_urls.add(url)

    return sorted(article_urls)


article_urls = find_article_links(LISTING_URL)

print("พบ article URLs =", len(article_urls))
print("ตัวอย่าง 5 URLs:")
for u in article_urls[:5]:
    print(" -", u)
````

#### Step 4 — เปิดบทความทีละหน้าแล้วดึง “เนื้อหาที่ต้องการ”

ตอนนี้เรามี URL ของบทความแล้ว จึงเปิดทีละข่าวและแยก field ที่จะใช้วิเคราะห์ เช่น

* `url`
* `title`
* `published_at`
* `text`
* `word_count`

หลักสำคัญคือ **พยายามดึงเฉพาะตัวบทความ** ไม่ใช่ menu/footer/ข่าวแนะนำทั้งหน้า

ตัวอย่างใช้ `<article>` ก่อน ถ้าเว็บไซต์ไม่มี tag นี้จึง fallback ไปที่ `<main>` และนำข้อความจาก `<p>` มาต่อกัน  
เว็บไซต์อื่นอาจต้องเปลี่ยน selector เช่น `.article-body`, `.post-content` หรือ selector เฉพาะของเว็บนั้น


````python
# ===== Step 4: ดึงข้อมูลจากบทความ 1 หน้า =====

def extract_article(url):
    if not robots_allowed(url):
        print("robots.txt ไม่อนุญาต ข้าม:", url)
        return None

    r = session.get(url, timeout=20)
    r.raise_for_status()

    soup = BeautifulSoup(r.text, "html.parser")

    # ---- title ----
    h1 = soup.find("h1")
    title = h1.get_text(" ", strip=True) if h1 else ""

    # ---- published date: พยายามอ่านจาก meta หรือ time ----
    published_at = ""

    meta_time = soup.find("meta", attrs={"property": "article:published_time"})
    if meta_time and meta_time.get("content"):
        published_at = meta_time["content"]
    else:
        time_tag = soup.find("time")
        if time_tag:
            published_at = time_tag.get("datetime") or time_tag.get_text(" ", strip=True)

    # ---- article text ----
    container = soup.find("article") or soup.find("main")

    if container is None:
        return None

    paragraphs = []
    for p in container.find_all("p"):
        txt = p.get_text(" ", strip=True)

        # ตัดบรรทัดสั้น ๆ ที่มักไม่ใช่เนื้อหาหลัก
        if len(txt) >= 30:
            paragraphs.append(txt)

    text = " ".join(paragraphs).strip()

    # ถ้าดึงมาแล้วสั้นมาก มีโอกาสว่า selector ไม่ถูก
    if len(text) < 200:
        return None

    return {
        "doc_id": normalize_url(url),
        "url": normalize_url(url),
        "title": title,
        "published_at": published_at,
        "text": text,
        "word_count": len(text.split()),
        "source_domain": urlparse(url).netloc,
    }
````

#### Step 5 — ให้ Robot ทำซ้ำกับหลายบทความ

นี่คือส่วนที่ทำให้มันเป็น “Robot”

แทนที่จะเปิดข่าวด้วยมือทีละหน้า โปรแกรมจะวน URL ที่หาได้ใน Step 3 แล้วเรียก `extract_article()` ซ้ำ ๆ

มี `time.sleep(DELAY)` ทุกครั้งเพื่อไม่ส่ง request ถี่เกินไป

ในช่วงพัฒนาให้เริ่มจาก 5–20 ข่าวก่อน เมื่อดูตัวอย่างแล้วแน่ใจว่า `text` เป็นเนื้อหาข่าวจริง ค่อยเพิ่มจำนวน


````python
# ===== Step 5: เก็บหลายบทความอัตโนมัติ =====

def collect_articles(article_urls, max_articles=20, delay=1.5):
    rows = []

    for i, url in enumerate(article_urls[:max_articles], start=1):
        try:
            item = extract_article(url)

            if item is not None:
                rows.append(item)
                print(f"[{i}] OK   :", item["title"][:70])
            else:
                print(f"[{i}] SKIP :", url)

        except Exception as e:
            print(f"[{i}] ERROR:", url, e)

        time.sleep(delay)

    return rows


rows = collect_articles(
    article_urls,
    max_articles=MAX_ARTICLES,
    delay=DELAY
)

print("\nเก็บบทความที่ใช้ได้ =", len(rows))
````

#### Step 6 — ตรวจว่า “1 row = 1 ข่าว” จริงหรือไม่

อย่าเชื่อเพียงว่าโค้ดรันผ่าน ต้องเปิดดูตัวอย่างข้อมูลด้วย

สิ่งที่ต้องตรวจ:
1. URL แต่ละแถวเป็น article URL จริงหรือไม่
2. `title` ตรงกับข่าวหรือไม่
3. `text` เป็นเนื้อหาข่าว ไม่ใช่ menu/footer
4. มีข่าวซ้ำหรือไม่
5. ข้อความสั้นผิดปกติหรือไม่

หลังผ่านขั้นนี้จึงค่อยบันทึกเป็น corpus และนำไป Text Analytics


````python
# ===== Step 6: Quality Check =====

df_robot = pd.DataFrame(rows)

if len(df_robot) > 0:
    print("จำนวนก่อน clean =", len(df_robot))
    print("URL ซ้ำ =", df_robot["url"].duplicated().sum())
    print("ข้อความซ้ำ =", df_robot["text"].duplicated().sum())

    # ตัด URL ซ้ำและข้อความซ้ำ
    df_robot = (
        df_robot
        .drop_duplicates(subset=["url"])
        .drop_duplicates(subset=["text"])
        .copy()
    )

    # ตัวอย่างเกณฑ์ง่าย ๆ: ข่าวต้องมีอย่างน้อย 80 คำ
    df_robot = df_robot[df_robot["word_count"] >= 80].copy()

    print("จำนวนหลัง clean =", len(df_robot))

    display(
        df_robot[
            ["title", "published_at", "word_count", "url"]
        ].head(10)
    )

    # อ่านเนื้อหาจริง 1 ข่าวเพื่อตรวจด้วยตา
    print("\nตัวอย่าง text:\n")
    print(df_robot.iloc[0]["text"][:1500])
else:
    print("ยังไม่มีข้อมูล ตรวจ article pattern / selector / robots.txt")
````
