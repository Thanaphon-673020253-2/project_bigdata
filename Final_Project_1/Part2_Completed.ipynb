## เทคนิคที่อาจใช้ — เลือกเท่าที่จำเป็น

ไม่ต้องใช้ทั้งหมด โดยทั่วไป **2–4 เทคนิคก็เพียงพอ** ให้เลือกตามคำถามของทีม

| เทคนิค | เหมาะเมื่ออยากรู้ |
|---|---|
| **Word / n-gram frequency** | คนพูดถึงอะไรบ่อย |
| **TF-IDF / distinctive words** | สองกลุ่มพูดต่างกันอย่างไร |
| **Sentiment** | ข้อความเป็นบวก/ลบเพียงใด |
| **Aspect analysis** | คนพูดถึงราคา คุณภาพ การส่งของ ฯลฯ อย่างไร |
| **Topic modeling** | มีประเด็นอะไรที่เราอาจยังไม่รู้ |
| **Trend / segmentation** | ปัญหาเปลี่ยนตามเวลา หรือหนักในกลุ่มไหน |

> เทคนิคขั้นสูง เช่น embedding, clustering, regression, classification หรือ anomaly detection ทำได้ แต่ **ไม่บังคับ** และเหมาะสำหรับ Bonus เมื่อใช้ถูกและอธิบายได้

ถ้ายังไม่เคยใช้เทคนิคใด ให้ทดลองกับข้อมูลชุดเล็กก่อน แล้วตรวจด้วยตาว่าผลสมเหตุสมผลหรือไม่

> **หลักฐาน:** ตัวเลขพร้อม n และชี้ว่าดูได้จากกราฟหรือตารางไหน \

> **ความมั่นใจ:** สูง กลาง หรือ ต่ำ พร้อมเหตุผลสั้น ๆ

**ตัวอย่างที่ยังไม่ใช่ insight**

````python
import json
import random
import pandas as pd

# กำหนดจำนวนตัวอย่างที่ต้องการสุ่ม และล็อค Seed เพื่อให้ผลการสุ่มคงที่
N = SAMPLE_N
random.seed(RANDOM_SEED)

samples = []      # ลิสต์สำหรับเก็บข้อมูลตัวอย่าง (Reservoir)
valid_count = 0  # นับจำนวนบรรทัดที่แปลงเป็น JSON สำเร็จ (ใช้คำนวณโอกาสสุ่ม)

# เปิดไฟล์อ่านทีละบรรทัดเพื่อประหยัดหน่วยความจำ (RAM)
with open(review_path, "r", encoding="utf-8") as f:
    for line in f:
        try:
            # แปลงข้อความ JSON ในบรรทัดนั้นให้เป็น Dictionary
            record = json.loads(line)
            valid_count += 1  # นำนับเฉพาะบรรทัดที่อ่านได้สำเร็จ

            # --- Reservoir Sampling Algorithm ---
            # 1. หากเก็บข้อมูลยังไม่ครบ N ให้ใส่ลงในลิสต์ทันที
            if len(samples) < N:
                samples.append(record)
            else:
                # 2. หากเกิน N ให้สุ่มตำแหน่ง j ตั้งแต่ 0 ถึง valid_count - 1
                j = random.randint(0, valid_count - 1)

                # ถ้าค่าที่สุ่มได้ตกอยู่ในช่วง 0 ถึง N - 1 ให้สุ่มแทนที่ข้อมูลเก่าในตำแหน่งนั้น
                if j < N:
                    samples[j] = record

        except json.JSONDecodeError:
            # ข้ามบรรทัดที่รูปแบบ JSON เสีย/ไม่ถูกต้อง โดยไม่นำมาคำนวณ
            continue

# แปลงลิสต์ตัวอย่างที่สุ่มได้ให้กลายเป็น Pandas DataFrame
df_review = pd.DataFrame(samples)

# แสดงจำนวนรายการทั้งหมดที่สุ่มมาได้
print("จำนวน Review ที่สุ่มได้:", len(df_review))
````

````python
needed_parent_asin = set(
    df_review["parent_asin"]
    .dropna()
    .astype(str)
)

print(
    "จำนวน parent_asin ที่ต้องหา:",
    len(needed_parent_asin)
)
````

````python
#ดึงข้อมูล product
product_records = []
bad_meta_lines = 0

with open(meta_path, "r", encoding="utf-8") as f:

    for line_number, line in enumerate(f, start=1):

        try:
            product = json.loads(line)

            parent_asin = product.get("parent_asin")

            if parent_asin is not None:
                parent_asin = str(parent_asin)

            if parent_asin in needed_parent_asin:
                product_records.append(product)

        except json.JSONDecodeError:
            bad_meta_lines += 1
            continue

print("Products found:", len(product_records))
print("Bad metadata lines:", bad_meta_lines)
````

````python
df_product = pd.DataFrame(product_records)

print(df_product.shape)

display(df_product.head())
````

````python
# แก้ข้อมูลที่ซ็ำ
df_product_unique = (
    df_product
    .drop_duplicates(
        subset="parent_asin",
        keep="first"
    )
    .copy()
)

print(
    "Products before:",
    len(df_product)
)

print(
    "Products after:",
    len(df_product_unique)
)
````

###Merge review and product

````python
df_merged = df_review.merge(
    df_product_unique,
    on="parent_asin",
    how="left",
    suffixes=("_review", "_product"),
    indicator=True,
    validate="many_to_one"
)
````

````python
#ตรวจสอบว่า merge ได้กี่ %
total = len(df_merged)

matched = (
    df_merged["_merge"] == "both"
).sum()

unmatched = (
    df_merged["_merge"] == "left_only"
).sum()

print(f"Total Reviews    : {total:,}")
print(f"Joined           : {matched:,}")
print(f"Not Joined       : {unmatched:,}")
print(f"Join Success     : {matched / total * 100:.2f}%")
print(f"Missing Metadata : {unmatched / total * 100:.2f}%")
print("Before Join:", len(df_review))
print("After Join :", len(df_merged))
if len(df_merged) == len(df_review):
    print("✓ จำนวนแถวไม่เพิ่มหลัง Join")
else:
    print("⚠ จำนวนแถวเพิ่มหลัง Join")
    print("เพิ่มขึ้น:", len(df_merged) - len(df_review))
````

### 2.3 สำรวจข้อมูลเท่าที่จำเป็น

> “สิ่งใดจาก EDA ทำให้ทีมอยากวิเคราะห์ต่อ และเพราะอะไร”

ไม่ต้องทำ EDA ทุกตัวแปร


````python

import os
import ast
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# ตั้งค่าสไตล์กราฟให้อ่านง่ายและสวยงาม
plt.style.use('seaborn-v0_8-whitegrid' if 'seaborn-v0_8-whitegrid' in plt.style.available else 'default')
plt.rcParams['font.family'] = 'sans-serif'
plt.rcParams['font.sans-serif'] = ['DejaVu Sans', 'Arial', 'Tahoma']

# โหลดข้อมูล
DATA_PATH = '/content/data.csv'
df = pd.read_csv(DATA_PATH)

# นิยามรีวิวเชิงลบ (1-2 ดาว) สำหรับ Brief 1 (ทีม Product Management)
df["negative_review"] = df["rating"] <= 2
df["word_count"] = df["text"].fillna("").astype(str).apply(lambda x: len(x.split()))

# สกัดชื่อหมวดหมู่สินค้าหลักจากคอลัมน์ categories
def get_product_category(cat_str):
    try:
        val = ast.literal_eval(cat_str)
        if isinstance(val, list) and len(val) > 0:
            return val[2] if len(val) > 2 else val[-1]
    except:
        pass
    return "Other / Unknown"

df["product_category"] = df["categories"].apply(get_product_category)

print(" โหลดข้อมูลสำเร็จจาก:", DATA_PATH)
print(f" จำนวนข้อมูลทั้งหมด: {len(df):,} รายการ")
print(f" รีวิวเชิงลบ (1–2 ดาว): {df['negative_review'].sum():,} รายการ ({df['negative_review'].mean()*100:.2f}%)")
````

````python
# ============================================================
# กราฟที่ 1: การแจกแจงคะแนนดาว (Rating Distribution - N = 30,000)
# เป้าหมาย Brief 1: ประเมินขนาดของปัญหารีวิวเชิงลบในภาพรวม
# ============================================================
fig, ax = plt.subplots(figsize=(8, 4.5))

rating_counts = df["rating"].value_counts().sort_index()
ratings = rating_counts.index.astype(int)
counts = rating_counts.values
colors = ['#d9534f', '#f0ad4e', '#5bc0de', '#428bca', '#5cb85c']

bars = ax.bar([f"{r} Stars" for r in ratings], counts, color=colors, width=0.55, edgecolor='#333333', linewidth=0.8)

for bar in bars:
    h = bar.get_height()
    pct = h / len(df) * 100
    ax.annotate(f"{h:,}\n({pct:.1f}%)",
                xy=(bar.get_x() + bar.get_width() / 2, h),
                xytext=(0, 4), textcoords="offset points",
                ha='center', va='bottom', fontsize=9.5, fontweight='bold')

total_neg = df["negative_review"].sum()
neg_pct = df["negative_review"].mean() * 100

ax.set_title(f"EDA 1: Distribution of Review Ratings (Negative 1-2 Stars: {total_neg:,} Reviews / {neg_pct:.1f}%)",
             fontsize=12, fontweight='bold', pad=14)
ax.set_xlabel("Review Rating", fontsize=11)
ax.set_ylabel("Number of Reviews", fontsize=11)
ax.set_ylim(0, max(counts) * 1.18)
plt.tight_layout()
plt.show()
````

**สิ่งที่ได้จาก EDA 1 (การแจกแจงคะแนนดาว):**

1. **ขนาดของปัญหาในภาพรวม:** จากข้อมูลพบว่าลูกค้าส่วนใหญ่ให้คะแนน 5 ดาว คิดเป็น 67.5% แต่ก็ยังมีรีวิวเชิงลบระดับ 1–2 ดาวรวมกันถึง **4,660 รายการ (15.60%)** โดยเป็นรีวิว 1 ดาวจำนวน 3,172 รายการ (10.60%) และรีวิว 2 ดาวจำนวน 1,488 รายการ (5.00%)

2. **ความเกี่ยวข้องกับ Brief 1:** จากจำนวนรีวิวเชิงลบที่มีค่อนข้างมาก แสดงให้เห็นว่าปัญหาความไม่พอใจของลูกค้าไม่ได้เกิดขึ้นแค่เป็นกรณีเล็ก ๆ หรือเฉพาะบางราย แต่เป็นปัญหาที่พบในข้อมูลอยู่พอสมควร ดังนั้นจึงมีเหตุผลที่จะนำปัญหาเหล่านี้ไปวิเคราะห์ต่อเพื่อดูว่าควรแก้ไขตรงไหนก่อน โดยเฉพาะเมื่อทีมมีงบประมาณ R&D จำกัด เราจึงต้องดูต่อใน **EDA 2** ว่ารีวิวเชิงลบเหล่านี้กระจุกตัวอยู่ที่กลุ่มสินค้าใดมากที่สุด


````python
# ============================================================
# กราฟที่ 2: เปรียบเทียบจำนวนรีวิวตามกลุ่มสินค้า (ครบ 100%)
# - รีวิวเชิงลบ (1-2 ดาว) | ปานกลาง (3 ดาว) | เชิงบวก (4-5 ดาว)
# เป้าหมาย Brief 1: ตอบคำถาม 'สินค้ากลุ่มไหนควรแก้ก่อนเป็นอันดับแรก'
# ============================================================
df["neg_review"] = df["rating"] <= 2
df["neu_review"] = df["rating"] == 3
df["pos_review"] = df["rating"] >= 4

cat_summary = df.groupby("product_category").agg(
    total_reviews=("rating", "count"),
    neg_reviews=("neg_review", "sum"),
    neg_rate=("neg_review", "mean"),
    neu_reviews=("neu_review", "sum"),
    neu_rate=("neu_review", "mean"),
    pos_reviews=("pos_review", "sum"),
    pos_rate=("pos_review", "mean")
).reset_index()

cat_summary["neg_pct"] = (cat_summary["neg_rate"] * 100).round(2)
cat_summary["neu_pct"] = (cat_summary["neu_rate"] * 100).round(2)
cat_summary["pos_pct"] = (cat_summary["pos_rate"] * 100).round(2)

# คัดกรองกลุ่มสินค้า Top 10 และเรียงตาม 'จำนวนรีวิวเชิงลบ'
top10_cat = cat_summary[cat_summary["total_reviews"] >= 100].sort_values(by="neg_reviews", ascending=True).tail(10).reset_index(drop=True)

fig, ax = plt.subplots(figsize=(12, 6.2))
y_positions = np.arange(len(top10_cat))

b_neg = ax.barh(y_positions, top10_cat["neg_pct"], color='#d9534f', edgecolor='#333333', linewidth=0.6, label="Negative (1-2 Stars)")
b_neu = ax.barh(y_positions, top10_cat["neu_pct"], left=top10_cat["neg_pct"], color='#f0ad4e', edgecolor='#333333', linewidth=0.6, label="Neutral (3 Stars)")
b_pos = ax.barh(y_positions, top10_cat["pos_pct"], left=top10_cat["neg_pct"] + top10_cat["neu_pct"], color='#5cb85c', edgecolor='#333333', linewidth=0.6, label="Positive (4-5 Stars)")

for i, row in top10_cat.iterrows():
    ax.text(row["neg_pct"] / 2, y_positions[i], f"{int(row['neg_reviews']):,}\n({row['neg_pct']:.1f}%)",
            ha='center', va='center', color='white', fontweight='bold', fontsize=8.5)
    if row["neu_pct"] >= 4.5:
        ax.text(row["neg_pct"] + row["neu_pct"] / 2, y_positions[i], f"{int(row['neu_reviews']):,}\n({row['neu_pct']:.1f}%)",
                ha='center', va='center', color='black', fontweight='bold', fontsize=8)
    ax.text(row["neg_pct"] + row["neu_pct"] + row["pos_pct"] / 2, y_positions[i], f"{int(row['pos_reviews']):,}\n({row['pos_pct']:.1f}%)",
            ha='center', va='center', color='white', fontweight='bold', fontsize=8.5)

ax.set_yticks(y_positions)
ax.set_yticklabels(top10_cat["product_category"], fontsize=10)
ax.set_xlabel("Review Proportion (% of Category Total - Total = 100%)", fontsize=11)
ax.set_xlim(0, 100)
ax.set_title("EDA 2: Breakdown of Negative, Neutral, and Positive Reviews by Category (Sorted by Negative Count)", fontsize=12.5, fontweight='bold', pad=14)
ax.legend(loc="upper center", bbox_to_anchor=(0.5, -0.1), ncol=3, fontsize=10)
plt.tight_layout()
plt.show()
````

> **สิ่งที่ได้จาก EDA 2**
>
> จากกราฟนี้เราลองดูว่า ในแต่ละกลุ่มสินค้ามีรีวิวเชิงลบมากน้อยแค่ไหน แล้วพบว่า **Printers & Accessories** เป็นกลุ่มที่มีรีวิวเชิงลบสูงที่สุด โดยมีทั้งหมด **975 รีวิว หรือประมาณ 24.7%** ของรีวิวในกลุ่มนี้
>
> พอเห็นตัวเลขนี้ เราเลยเลือกกลุ่ม **`Printers & Accessories`** มาดูต่อ เพราะมีทั้งจำนวนรีวิวที่เยอะและสัดส่วนรีวิวเชิงลบค่อนข้างสูง ทำให้มองเห็นว่ากลุ่มนี้น่าจะมีปัญหาที่ควรเข้าไปดูให้ละเอียดมากกว่ากลุ่มอื่น
>
> ดังนั้นในขั้นต่อไป เราจะเจาะลึกลงไปอีกว่า **ในกลุ่ม `Printers & Accessories` ปัญหาจริง ๆ ที่ลูกค้าพูดถึงคืออะไร และสินค้าหรือเรื่องไหนที่ควรแก้ก่อน** ซึ่งจะนำไปวิเคราะห์ต่อใน EDA 3


````python
# ============================================================
# กราฟที่ 3: จำนวนรีวิวเชิงลบแยกตามสายผลิตภัณฑ์ในหมวดเครื่องพิมพ์ (Printers & Accessories)
# เป้าหมาย Brief 1: เจาะลึกหา 'สินค้าตัวการ' ที่ลูกค้าบ่นมากที่สุดเพื่อจัดสรรงบ R&D ให้ตรงจุด
# ============================================================
printers_df = df[df["product_category"] == "Printers & Accessories"].copy()

def get_leaf_category(cat_str):
    try:
        val = ast.literal_eval(cat_str)
        if isinstance(val, list) and len(val) > 0:
            return val[-1]
    except:
        pass
    return "Other Printers"

printers_df["leaf_category"] = printers_df["categories"].apply(get_leaf_category)

# รวมจำนวนรีวิวเชิงลบและรีวิวทั้งหมดตามสายผลิตภัณฑ์
sub_summary = printers_df.groupby("leaf_category").agg(
    total_reviews=("rating", "count"),
    neg_reviews=("neg_review", "sum")
).reset_index()

sub_summary["neg_pct"] = (sub_summary["neg_reviews"] / sub_summary["total_reviews"] * 100).round(1)

# คัดกรองกลุ่มที่มีรีวิว >= 50 และเรียงลำดับจาก 'จำนวนรีวิวเชิงลบ' มากที่สุดลงไปน้อยที่สุด (Top-down)
top_sub = sub_summary[sub_summary["total_reviews"] >= 50].sort_values(by="neg_reviews", ascending=False).reset_index(drop=True)

fig, ax = plt.subplots(figsize=(10, 4.8))
y_pos = np.arange(len(top_sub))

# ใช้สีแดงเน้นกลุ่มที่มีปัญหาเกิน 500 รีวิว และสีส้ม/ฟ้าสำหรับกลุ่มรองลงมา
colors = ['#d9534f' if r >= 500 else ('#e67e22' if r >= 100 else '#428bca') for r in top_sub["neg_reviews"]]
bars = ax.barh(y_pos, top_sub["neg_reviews"], color=colors, height=0.55, edgecolor='#333333', linewidth=0.7)

ax.set_yticks(y_pos)
ax.set_yticklabels(top_sub["leaf_category"], fontsize=10, fontweight='bold')
ax.invert_yaxis()  # ให้อันดับ 1 อยู่ด้านบนสุด
ax.set_xlabel("Number of Negative Reviews (1-2 Stars)", fontsize=11, fontweight='bold')
ax.set_title("EDA 3: Negative Review Counts by Product Line in Printers & Accessories", fontsize=12.5, fontweight='bold', pad=14)
ax.set_xlim(0, max(top_sub["neg_reviews"]) * 1.25)
ax.grid(True, linestyle='--', alpha=0.5, axis='x')

for b, pct, tot in zip(bars, top_sub["neg_pct"], top_sub["total_reviews"]):
    w = b.get_width()
    ax.annotate(f"{int(w):,} reviews ({pct:.1f}% of {tot:,})",
                xy=(w, b.get_y() + b.get_height() / 2),
                xytext=(6, 0), textcoords='offset points',
                ha='left', va='center', fontsize=9.5, fontweight='bold', color='#333333')

plt.tight_layout()
plt.show()
````

> **สิ่งที่ได้จาก EDA 3 (เจาะลึกสายผลิตภัณฑ์ในหมวดเครื่องพิมพ์):**
> 1. **สินค้าที่ลูกค้าบ่นมากที่สุด (Primary Culprit):** ภายในหมวดเครื่องพิมพ์ พบว่า **`Inkjet Ink Cartridges` (ตลับหมึกพิมพ์อิงค์เจ็ท)** มีรีวิวเชิงลบสูงเป็นอันดับ 1 ถึง **513 รายการ (52.6% ของรีวิวเชิงลบในหมวดเครื่องพิมพ์ทั้งหมด)** และตลับผงหมึก Toner (149 รายการ) และเครื่องพิมพ์แทงค์ Ink Tank (116 รายการ)
> 2. **ตอบคำถาม Brief 1:** ทีม R&D ได้เป้าหมายที่ชัดเจนว่า **"ตลับหมึกอิงค์เจ็ท (Cartridges)"** คือ Quick Win อันดับ 1 ที่ต้องจัดสรรงบประมาณเข้าไปปรับปรุงสูตรหมึกและตลับหมึก เพื่อลดปริมาณข้อร้องเรียนของลูกค้าได้เกินครึ่งหนึ่งของทั้งหมวดเครื่องพิมพ์ในทันที

### 2.4 เตรียมข้อความและสร้าง Text Features

ก่อนนำข้อความไปวิเคราะห์ จำเป็นต้องมีการ **preprocessing** เพื่อทำให้ข้อความอยู่ในรูปที่เหมาะสมกับเทคนิคที่จะใช้ อย่างไรก็ตาม ในงานนี้ไม่จำเป็นต้องทำ preprocessing ทุกขั้นตอน ให้เลือกทำ **เท่าที่จำเป็น** ตามวัตถุประสงค์ของการวิเคราะห์

ตัวอย่างเช่น การเปลี่ยนข้อความเป็นตัวพิมพ์เล็ก (lower case) การนำ URL หรือ HTML ออก การแบ่งข้อความเป็นคำ (tokenization) หรือการตัด stopwords เช่น `the`, `and`, `is` ออก ในกรณีที่คำเหล่านี้ไม่ได้ช่วยในการวิเคราะห์

**ไม่บังคับให้ใช้ stemming หรือ lemmatization** หากไม่มีเหตุผลว่าขั้นตอนดังกล่าวจะช่วยตอบคำถามของทีมได้ดีขึ้น

หลังจากเตรียมข้อความแล้ว ให้สร้าง **Text Features อย่างน้อย 2 ชนิด** เพื่อนำข้อความมาเปลี่ยนให้อยู่ในรูปข้อมูลที่สามารถนำไปสำรวจ เปรียบเทียบ หรือวิเคราะห์ต่อได้

---

#### Text Feature คืออะไร?

Text Feature คือ **ข้อมูลเชิงตัวเลขที่สกัดหรือสร้างขึ้นจากข้อความ** เพื่อแทนลักษณะหรือความหมายบางอย่างของข้อความนั้น

Text Feature อาจอยู่ในรูปของ

* **ค่าตัวเลขค่าเดียว** เช่น จำนวนคำ หรือ sentiment score
* **ตัวแปรหลายตัว** เช่น `price=1`, `quality=0`, `delivery=1`
* **เวกเตอร์หลายมิติ** เช่น TF-IDF vector หรือ Text Embedding

ตัวอย่างเช่น รีวิว

> `"The product is easy to use but the delivery was very slow."`

สามารถเปลี่ยนเป็น feature ได้หลายรูปแบบ

| Feature                    |                 ตัวอย่างค่าที่ได้ |
| -------------------------- | --------------------------------: |
| จำนวนคำในรีวิว             |                                12 |
| Sentiment score            |                             -0.15 |
| กล่าวถึง usability หรือไม่ |                                 1 |
| กล่าวถึง delivery หรือไม่  |                                 1 |
| กล่าวถึง price หรือไม่     |                                 0 |
| TF-IDF vector              |   `[0.00, 0.41, 0.00, 0.72, ...]` |
| Embedding vector           | `[0.21, -0.08, 0.63, ..., -0.14]` |

ดังนั้นข้อความหนึ่งข้อความสามารถถูกแทนได้ทั้งด้วย **feature ที่ตีความได้โดยตรง** และ **feature แบบเวกเตอร์ที่ใช้แทนเนื้อหาหรือความหมายของข้อความ**

---

## ตัวอย่าง Text Features ที่สามารถเลือกใช้

ผมจะแก้ตารางเดิมเป็นแบบนี้:

| Text Feature          | คืออะไร                                     | ตัวอย่าง                     | ใช้ตอบคำถามอะไรได้                                    |
| --------------------- | ------------------------------------------- | ---------------------------- | ----------------------------------------------------- |
| **Review Length**     | จำนวนคำหรือตัวอักษร                         | 120 words                    | รีวิวที่ยาวกว่าได้รับ helpful vote มากกว่าหรือไม่     |
| **Sentiment Score**   | คะแนนความเป็นบวก/ลบ                         | `-0.72`, `+0.85`             | ความรู้สึกลูกค้าสัมพันธ์กับ rating หรือไม่            |
| **Aspect Indicators** | ระบุว่าพูดถึงหัวข้อใด                       | `delivery=1`, `price=0`      | ลูกค้าบ่นเรื่องใดมากที่สุด                            |
| **Lexical Diversity** | ความหลากหลายของคำ                           | unique words / total words   | รีวิวที่เขียนละเอียดมีประโยชน์มากกว่าหรือไม่          |
| **TF-IDF Vector**     | เวกเตอร์ที่ให้น้ำหนักคำเด่นในเอกสาร         | `[0, .35, 0, .81, ...]`      | คำใดแยกกลุ่มรีวิวบวกกับลบ                             |
| **Text Embedding**    | dense vector ที่แทนความหมายโดยรวมของข้อความ | `[.12, -.44, .08, ...]`      | รีวิวใดมีความหมายคล้ายกัน หรือมี customer themes อะไร |
| **Custom Feature**    | feature ที่ทีมสร้างตามโจทย์                 | `refund=1`, `defect_count=3` | ตอบ business question เฉพาะของทีม                     |

````python
# ============================================================
# 2.4.1 Text Preprocessing
# ============================================================
import re
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

def clean_text(text):
    """
    ทำความสะอาดข้อความรีวิวเท่าที่จำเป็น:
    - จัดการค่า missing
    - แปลงเป็นตัวพิมพ์เล็ก
    - ลบ HTML tags
    - ลบ URLs
    - จัดการช่องว่างซ้ำซ้อน
    """
    if pd.isna(text):
        return ""

    text = str(text).lower()
    text = re.sub(r"<[^>]+>", " ", text)
    text = re.sub(r"https?://\S+|www\.\S+", " ", text)
    text = re.sub(r"\s+", " ", text).strip()

    return text


# สร้าง Clean Text สำหรับใช้ใน Aspect Analysis
df["clean_text"] = df["text"].apply(clean_text)

# เก็บข้อความอีกชุดไว้สำหรับ Sentiment
# เพื่อให้สามารถปรับการ preprocess แยกจาก Aspect ได้ในอนาคต
df["sentiment_text"] = df["text"].fillna("").astype(str)

# ข้อความสำหรับ Keyword Matching
df["aspect_text"] = df["clean_text"]


print("ทำความสะอาดข้อความรีวิวสำเร็จ")
print(f"จำนวนข้อความทั้งหมด: {len(df):,} รายการ")
print(
    f"จำนวนข้อความที่เป็นค่าว่างหลังทำความสะอาด: "
    f"{(df['clean_text'] == '').sum():,} รายการ"
)

sample_comp = (
    df[["text", "clean_text"]]
    .dropna()
    .head(5)
    .copy()
)

sample_comp.columns = [
    "Original Text (ข้อความเดิม)",
    "Cleaned Text (ข้อความที่ทำความสะอาดแล้ว)"
]

display(sample_comp)
````

````python
# ============================================================
# 2.4.2 Text Feature 1: Sentiment Score (VADER Lexicon Analyzer)
# คำถามธุรกิจ: ความรู้สึกลูกค้าสัมพันธ์กับ rating หรือไม่
# ============================================================
import sys
import os
venv_site = r"W:\Learn_Program\project_bigdata\.venv\Lib\site-packages"
if os.path.exists(venv_site) and venv_site not in sys.path:
    sys.path.insert(0, venv_site)

from nltk.sentiment.vader import SentimentIntensityAnalyzer
import nltk
try:
    nltk.data.find('sentiment/vader_lexicon.zip')
except LookupError:
    nltk.download('vader_lexicon', quiet=True)

sia = SentimentIntensityAnalyzer()

# 1. คำนวณ Compound Sentiment Score [-1.0 ถึง +1.0]
df["sentiment_score"] = df["clean_text"].apply(lambda t: sia.polarity_scores(t)["compound"])

def label_sentiment(score):
    if score >= 0.05:
        return "Positive"
    elif score <= -0.05:
        return "Negative"
    else:
        return "Neutral"

df["sentiment_label"] = df["sentiment_score"].apply(label_sentiment)

# 2. คำนวณค่าความสัมพันธ์ Correlation
corr = df["rating"].corr(df["sentiment_score"])

# 3. ตารางสรุป Sentiment ตามระดับ Rating
sentiment_by_rating = df.groupby("rating").agg(
    จำนวนรีวิว=("rating", "count"),
    Sentiment_เฉลี่ย=("sentiment_score", "mean"),
    Sentiment_มัธยฐาน=("sentiment_score", "median"),
    สัดส่วนเชิงลบ_pct=("sentiment_label", lambda x: (x == "Negative").mean() * 100),
    สัดส่วนเชิงบวก_pct=("sentiment_label", lambda x: (x == "Positive").mean() * 100)
).round(3).reset_index()

sentiment_by_rating.columns = [
    "คะแนน (Rating)", "จำนวนรีวิว", "Sentiment เฉลี่ย", "Sentiment มัธยฐาน",
    "สัดส่วน Negative (%)", "สัดส่วน Positive (%)"
]

print(f" ค่าสหสัมพันธ์ (Pearson Correlation) ระหว่าง Rating และ Sentiment Score: {corr:.4f}")
display(sentiment_by_rating)

# 4. กราฟเดี่ยวแสดงค่าเฉลี่ย Sentiment Score พร้อมระบุที่มาการคำนวณและจำนวนตัวอย่าง (n) ชัดเจน
fig, ax = plt.subplots(figsize=(9.5, 5.2))

ratings_labels = [f"{int(r)} Stars" for r in sentiment_by_rating["คะแนน (Rating)"]]
scores = sentiment_by_rating["Sentiment เฉลี่ย"].values
counts = sentiment_by_rating["จำนวนรีวิว"].values
colors = ['#d9534f', '#f0ad4e', '#5bc0de', '#428bca', '#5cb85c']

bars = ax.bar(ratings_labels, scores, color=colors, width=0.52, edgecolor='#333333', linewidth=0.7)
ax.axhline(0, color='gray', linestyle='--', linewidth=1.2, alpha=0.7)

for b, s, cnt in zip(bars, scores, counts):
    h = b.get_height()
    ypos = h + 0.04 if h >= 0 else h - 0.14
    va = 'bottom' if h >= 0 else 'top'
    tone = 'Negative' if s < 0 else ('Neutral' if s < 0.15 else 'Positive')
    lbl = f"{s:+.2f} ({tone})\n(n = {cnt:,})"
    ax.annotate(lbl, xy=(b.get_x() + b.get_width() / 2, ypos),
                ha='center', va=va, fontsize=9, fontweight='bold')

ax.set_title(f"Feature 1: Customer Sentiment Score by Rating (Correlation r = +{corr:.2f})\n"
             "Calculated as: Mean VADER Compound Score of review texts in each star group (-1.0 to +1.0)",
             fontsize=11.5, fontweight='bold', pad=14)

ax.set_xlabel("Review Rating", fontsize=11, fontweight='bold')
ax.set_ylabel("Mean Sentiment Score (-1.0 Negative to +1.0 Positive)", fontsize=10.5, fontweight='bold')
ax.set_ylim(-0.48, 0.88)
ax.grid(True, linestyle='--', alpha=0.5, axis='y')

ax.text(4.4, 0.78, "Positive Tone Zone", color='#2e7d32', fontweight='bold', fontsize=9.5, ha='right')
ax.text(0.4, -0.40, "Negative Tone Zone", color='#c62828', fontweight='bold', fontsize=9.5, ha='left')

plt.tight_layout()
plt.show()
````

**ข้อสังเกตและสิ่งที่ได้จาก Feature 1 (Sentiment Score):**

* **1. การคำนวณและที่มา:** คำนวณด้วย **VADER Lexicon Analyzer** จากข้อความรีวิว (`clean_text`) โดยสกัดค่า Compound Score ซึ่งอยู่ในช่วง **[-1.0 ถึง +1.0]** จากนั้นนำมาหาค่าเฉลี่ยตามระดับดาว

* **2. ตอบคำถามธุรกิจ: “ความรู้สึกลูกค้าสัมพันธ์กับ Rating หรือไม่”**

  * ผลการวิเคราะห์พบว่า **Rating และ Sentiment Score มีความสัมพันธ์เชิงบวก** โดยค่า Pearson Correlation เท่ากับ **r = +0.5476**
  * หมายความว่า โดยทั่วไปรีวิวที่ให้คะแนนดาวสูงจะมี Sentiment Score สูงขึ้นตาม ขณะที่รีวิวที่ให้คะแนนต่ำมักมี Sentiment Score ต่ำกว่า

* **3. ข้อสังเกตเชิงลึกสำหรับทีม Product Management:**

  * รีวิว 1–2 ดาวบางรายการอาจมี Sentiment Score เป็นกลางหรือเป็นบวกเล็กน้อย แม้จะกล่าวถึงปัญหาของสินค้า เนื่องจากผู้รีวิวอาจอธิบายอาการเสียในลักษณะเชิงเทคนิคโดยไม่ได้ใช้ถ้อยคำที่แสดงอารมณ์รุนแรง
  * ดังนั้น **Sentiment Score เพียงอย่างเดียวไม่เพียงพอสำหรับระบุปัญหาของสินค้า** และควรใช้ร่วมกับ Aspect Indicators เพื่อดูว่าลูกค้ากำลังกล่าวถึงประเด็นใด


````python
# ============================================================
# 2.4.3 Text Feature 2: Aspect Indicators (Binary Topic Features)
# คำถามธุรกิจ: ลูกค้าบ่นเรื่องใดมากที่สุด
# ============================================================
aspects_dict = {
    "functionality": [
        "works", "work", "working", "print", "printing", "jam", "jammed",
        "ink", "cartridge", "connect", "connection", "setup", "software", "error"
    ],
    "quality": [
        "quality", "durable", "durability", "sturdy", "flimsy", "broke",
        "broken", "cheap", "crack", "cracked", "defect", "defective", "poor quality"
    ],
    "service": [
        "service", "support", "refund", "return", "returned", "warranty", "replace", "replacement"
    ],
    "price": [
        "price", "expensive", "cost", "value", "worth", "money", "affordable"
    ],
    "delivery": [
        "delivery", "shipping", "shipped", "arrive", "arrived", "package", "packaging", "box"
    ]
}

# สร้าง Binary Features (0 หรือ 1) สำหรับแต่ละ Aspect
aspect_cols = []
for aspect, keywords in aspects_dict.items():
    col_name = f"aspect_{aspect}"
    aspect_cols.append(col_name)
    pattern = r"\b(?:" + "|".join(re.escape(k) for k in keywords) + r")\b"
    df[col_name] = df["clean_text"].str.contains(pattern, regex=True, na=False).astype(int)

# 1. ตารางเปรียบเทียบสัดส่วนการกล่าวถึงใน รีวิวทั้งหมด vs รีวิวเชิงลบ (1-2 ดาว)
df_neg = df[df["rating"] <= 2]
total_neg_count = len(df_neg)

aspect_comparison = pd.DataFrame({
    "Aspect (ประเด็น)": [
        "Functionality (การทำงาน/ฟังก์ชัน/กลไก)",
        "Quality (คุณภาพ/ความคงทนของวัสดุ)",
        "Service (บริการ/การคืนของ)",
        "Price (ราคา/ความคุ้มค่า)",
        "Delivery (การจัดส่ง/พัสดุ)"
    ],
    "Aspect_Clean": [
        "Functionality",
        "Quality",
        "Service",
        "Price",
        "Delivery"
    ],
    "จำนวนรีวิวเชิงลบที่บ่น": [df_neg[c].sum() for c in aspect_cols],
    "Mention Rate รีวิวเชิงลบ (%)": (df_neg[aspect_cols].mean() * 100).round(2).values,
    "Mention Rate รวม (%)": (df[aspect_cols].mean() * 100).round(2).values,
    "คะแนนเฉลี่ยเมื่อพูดถึง": [df[df[c] == 1]["rating"].mean() for c in aspect_cols],
    "Negative Rate (%)": [df[df[c] == 1]["negative_review"].mean() * 100 for c in aspect_cols]
}).round(2).sort_values(by="Mention Rate รีวิวเชิงลบ (%)", ascending=False)

display(aspect_comparison[["Aspect (ประเด็น)", "จำนวนรีวิวเชิงลบที่บ่น", "Mention Rate รีวิวเชิงลบ (%)", "Mention Rate รวม (%)", "คะแนนเฉลี่ยเมื่อพูดถึง"]].reset_index(drop=True))

# 2. กราฟเดี่ยวแสดงเฉพาะชื่อ Aspect อ่านง่าย สะอาดตา และระบุที่มาการคำนวณชัดเจน
fig, ax = plt.subplots(figsize=(9.5, 4.5))
y_pos = np.arange(len(aspect_comparison))
pcts = aspect_comparison["Mention Rate รีวิวเชิงลบ (%)"].values
cnts = aspect_comparison["จำนวนรีวิวเชิงลบที่บ่น"].values
aspect_names = aspect_comparison["Aspect_Clean"].values
colors_f2 = ['#d9534f', '#e67e22', '#f39c12', '#f1c40f', '#95a5a6']

bars = ax.barh(y_pos, pcts, color=colors_f2, height=0.52, edgecolor='#333333', linewidth=0.7)
ax.set_yticks(y_pos)
ax.set_yticklabels(aspect_names, fontsize=11, fontweight='bold')
ax.invert_yaxis()
ax.set_xlabel("Percentage of Negative Reviews Mentioning Issue (%)", fontsize=11, fontweight='bold')
ax.set_title("Feature 2: What Do Customers Complain About Most? (Negative Reviews 1-2 Stars)\n"
             f"Calculated as: (Reviews Mentioning Issue / Total {total_neg_count:,} Negative Reviews) x 100",
             fontsize=12, fontweight='bold', pad=14)

ax.set_xlim(0, 50)
ax.grid(True, linestyle='--', alpha=0.5, axis='x')

for i, (b, pct, c) in enumerate(zip(bars, pcts, cnts)):
    w = b.get_width()
    ax.annotate(f"{pct:.1f}%  ({c:,} / {total_neg_count:,} reviews)", xy=(w, b.get_y() + b.get_height() / 2),
                xytext=(6, 0), textcoords='offset points', ha='left', va='center', fontsize=9.5, fontweight='bold', color='#333333')

plt.tight_layout()
plt.show()
````

**ข้อสังเกตและสิ่งที่ได้จาก Feature 2 (Aspect Indicators):**

* **1. การคำนวณและที่มา:** ใช้ Keyword Matching ตามพจนานุกรมคำศัพท์ 5 ด้าน เพื่อสร้าง Binary Indicators โดย `1` หมายถึงพบคำที่อยู่ในกลุ่มประเด็นนั้น และ `0` หมายถึงไม่พบคำดังกล่าว

* **2. ผลการวิเคราะห์จาก Negative Reviews ทั้งหมด 4,660 รีวิว**

  1. **Functionality:** 34.9% หรือ 1,625 รีวิว
  2. **Quality:** 22.1% หรือ 1,028 รีวิว
  3. **Service:** 17.6% หรือ 819 รีวิว

* **4. ความเกี่ยวข้องกับ Client Brief 1:** ผลจากส่วนนี้ใช้เป็นภาพรวมของประเด็นที่ลูกค้ากล่าวถึงใน Negative Reviews ก่อนนำไปเจาะเฉพาะหมวด `Printers & Accessories` ในหัวข้อ 2.5 เพื่อจัดลำดับความสำคัญให้เหมาะกับโจทย์ธุรกิจ

### **รายงานสรุปหัวข้อ 2.4: การเตรียมข้อความและสร้าง Text Features**

#### **1. ทำ Preprocessing อะไร และทำไม**
* **เทคนิคที่เลือกใช้:**
  1. **Lowercasing (แปลงข้อความเป็นตัวพิมพ์เล็ก):** เพื่อลดความซ้ำซ้อนของคำ ทำให้คำเดียวกันที่มีตัวพิมพ์ใหญ่-เล็กต่างกัน (เช่น `Printer`, `printer`, `PRINTER`) ถูกมองเป็นคำเดียวกัน ช่วยให้การตรวจจับคำค้น (Keywords) ของ Aspect Indicators ทำงานได้แม่นยำ 100%
  2. **HTML & URL Removal (ตัดแท็ก HTML และลิงก์):** กำจัดขยะข้อความ เช่น `<br>`, `<p>`, `http://...` ที่ไม่มีข้อมูลความรู้สึก เพื่อลด Noise ก่อนคำนวณ Sentiment Analysis
  3. **Whitespace Normalization (จัดระเบียบช่องว่าง):** ตัดช่องว่างซ้ำซ้อนเพื่อให้ข้อความสะอาดพร้อมต่อการวิเคราะห์
* **เทคนิคที่ไม่ใช้และเหตุผลทางเทคนิค:**
  1. **ไม่ใช้ Stemming / Lemmatization:** เนื่องจากเครื่องมือ VADER ถูกออกแบบและเทรนมากับคำศัพท์เต็มรูปประโยคที่มีไวยากรณ์สมบูรณ์ (Inflected Forms) การตัดรากคำ (เช่น `broken` $\rightarrow$ `break`) จะทำลายคุณลักษณะทางภาษาและทำให้ Sentiment Score คลาดเคลื่อน
  2. **ไม่ใช้ Stopwords Removal:** เนื่องจากในกลุ่ม Stopwords มีคำปฏิเสธ (Negations เช่น `not`, `no`, `never`) และคำบอกระดับ (Intensifiers เช่น `very`, `too`) ซึ่งหากตัดออกจะทำให้ความหมายของรีวิวกลับด้านทันที (Negation Inversion เช่น *"not good"* จะกลายเป็น *"good"*)

---

#### **2. สร้าง Text Feature อะไรอย่างน้อย 2 ตัว**
1. **Feature 1: Sentiment Score (VADER Compound Score)**
   * เป็นตัวเลขทศนิยมต่อเนื่องอยู่ในช่วง **[-1.0, +1.0]** แทนระดับความรู้สึกของข้อความจากลบมากถึงบวกมาก
   * พร้อมจัดกลุ่ม `sentiment_label` แบ่งเป็น Positive ($\ge +0.05$), Neutral ($-0.05$ ถึง $+0.05$), Negative ($\le -0.05$)
2. **Feature 2: Aspect Indicators (Binary Indicators 0 หรือ 1)**
   * สร้างตัวแปร Dummy จำนวน 5 ด้าน: `aspect_functionality`, `aspect_quality`, `aspect_service`, `aspect_price`, `aspect_delivery` เพื่อระบุว่าลูกค้ารีวิวพูดถึงหัวข้อใดบ้าง

---

#### **3. Feature เหล่านั้นช่วยตอบคำถามทางธุรกิจอย่างไร (เชื่อมโยงกับ Client Brief 1)**
* **Sentiment Score ตอบว่า "ความรู้สึกลูกค้าสัมพันธ์กับ Rating หรือไม่":**
  * พบค่าความสัมพันธ์ **$r = +0.5476$** (สัมพันธ์กันเชิงบวกอย่างมีนัยสำคัญ) โดยรีวิว 5 ดาวได้ Sentiment เฉลี่ย **+0.598** ขณะที่รีวิว 1 ดาวได้ Sentiment เฉลี่ย **-0.150**
  * ยืนยันว่าคะแนนดาวสะท้อนความรู้สึกจริง และช่วยให้ตรวจจับรีวิวที่ให้ดาวปานกลางแต่เขียนบ่นด้วยข้อความเชิงลบได้อย่างแม่นยำ
* **Aspect Indicators ตอบว่า "ลูกค้าบ่นเรื่องใดมากที่สุด":**
  * ในกลุ่มลูกค้ารีวิวเชิงลบ (1–2 ดาว, 4,660 รีวิว) ลูกค้าบ่นเรื่อง **Functionality / การทำงานของสินค้า สูงสุดเป็นอันดับ 1 ถึง 34.9% (1,625 รีวิว)** และบ่นเรื่อง **Quality & Durability / ความคงทน เป็นอันดับ 2 ถึง 22.1% (1,028 รีวิว)**
  * ในขณะที่เรื่อง **Delivery & Packaging มีการบ่นเพียง 10.3% (481 รีวิว)**
* **บทสรุปเชิงกลยุทธ์สำหรับ Client Brief 1 (ส่งต่อสู่หัวข้อ 2.5):**
  * ทีม Product Management ต้องจัดสรรงบ R&D มุ่งเป้าไปที่ **การปรับปรุงกลไกการพิมพ์, การแก้ปัญหาหมึกตัน, และการเพิ่มความทนทานของตลับหมึกและเครื่องพิมพ์** เป็นลำดับแรก
  * ในหัวข้อ 2.5 ทีมจะนำ Features เหล่านี้ไปสร้างโมเดลจำลองเพื่อพิสูจน์ว่า หากแก้ไข 3 ปัญหาหลักนี้สำเร็จ คะแนนดาวเฉลี่ยจะเพิ่มขึ้นเท่าใด และจะลดปริมาณรีวิวเชิงลบลงได้กี่เปอร์เซ็นต์

### สรุปการเตรียมข้อความและ Text Features

ก่อนเริ่มวิเคราะห์ เราทำความสะอาดข้อความเท่าที่จำเป็น โดยแปลงข้อความเป็นตัวพิมพ์เล็ก ลบ HTML และ URL ที่ไม่เกี่ยวกับเนื้อหา และจัดการช่องว่างที่ซ้ำกัน เพื่อให้ข้อความอยู่ในรูปที่นำไปวิเคราะห์ต่อได้ง่ายขึ้น

เราไม่ได้ทำ Stemming หรือ Lemmatization เพราะงานนี้ต้องการดูคำที่ลูกค้าใช้จริงในรีวิว และเราใช้ Keyword สำหรับตรวจจับประเด็นปัญหาอยู่แล้ว ส่วน Stopwords ก็ไม่ได้ตัดออกทั้งหมด เพราะคำอย่าง “not”, “never” หรือ “very” มีผลต่อความหมายของประโยค โดยเฉพาะการวิเคราะห์ความรู้สึก

สำหรับ Text Features เราใช้ 2 กลุ่มหลัก คือ Sentiment Score และ Aspect Indicators

Sentiment Score ใช้ VADER เพื่อดูว่ารีวิวมีโทนไปทางบวกหรือลบมากแค่ไหน ผลที่ได้พบว่า Rating กับ Sentiment มีความสัมพันธ์เชิงบวกในระดับปานกลาง โดยมีค่า correlation เท่ากับ 0.5476 กล่าวง่าย ๆ คือ รีวิวที่ให้ดาวสูงมักมีข้อความที่เป็นบวกมากขึ้นตามไปด้วย

ส่วน Aspect Indicators ใช้ดูว่าลูกค้าพูดถึงประเด็นอะไรบ้าง จากรีวิว 1–2 ดาวในข้อมูลทั้งหมด พบว่า Functionality ถูกกล่าวถึงมากที่สุด รองลงมาคือ Quality, Service, Price และ Delivery อย่างไรก็ตาม ตัวเลขนี้เป็นการวัด “การกล่าวถึงประเด็น” จาก Keyword จึงไม่ได้หมายความว่า Keyword ทุกครั้งคือการร้องเรียนโดยตรง

ดังนั้น ในขั้นต่อไปเราจะเจาะเฉพาะหมวด Printers & Accessories เพื่อดูว่าในกลุ่มที่มีปัญหาสูง ประเด็นไหนถูกพูดถึงมากที่สุด และประเด็นไหนมีความเกี่ยวข้องกับคะแนนรีวิวต่ำมากที่สุด


### 2.5 หา Insight ที่ตอบ Client Brief

เลือกเทคนิคจากเมนูด้านบนตามความเหมาะสม **ไม่ต้องทำครบ**

ตัวอย่างแนวคิด:
* ลูกค้าที่ให้ 5 ดาวกับ 1–2 ดาวพูดต่างกันอย่างไร
* ปัญหาใดสัมพันธ์กับคะแนนต่ำ
* รีวิวที่มี helpful vote สูงเขียนต่างจากรีวิวทั่วไปหรือไม่
* มีปัญหาใดเพิ่มขึ้นตามเวลา
* ถ้าจำเป็น ค่อยแยกดูตามราคา สินค้าย่อย ช่วงเวลา หรือ verified purchase

> **Segmentation เป็นทางเลือก ไม่ใช่หัวข้อบังคับ** ทำเมื่อช่วยให้ insight ชัดขึ้นเท่านั้น


````python
# ============================================================
# 2.5.1 Problem Aspect Analysis
# เป้าหมาย Brief 1:
# "งบประมาณจำกัด แก้ได้ 3 เรื่อง ควรให้ความสำคัญกับเรื่องอะไร"
# ============================================================

printers = df[
    df["product_category"] == "Printers & Accessories"
].copy()

printers_neg = printers[
    printers["rating"] <= 2
].copy()

n_neg = len(printers_neg)
total_p = len(printers)

aspect_labels = {
    "aspect_functionality": "Functionality (Print/Ink/Cartridge)",
    "aspect_service": "Service & Warranty (Refund/Return)",
    "aspect_price": "Price & Value",
    "aspect_quality": "Quality & Durability",
    "aspect_delivery": "Delivery & Shipping"
}

rows = []

for col, label in aspect_labels.items():

    cnt_neg = printers_neg[col].sum()
    pct_neg = cnt_neg / n_neg * 100

    pct_all = printers[col].mean() * 100

    rating_when_mentioned = printers[
        printers[col] == 1
    ]["rating"].mean()

    rating_when_not_mentioned = printers[
        printers[col] == 0
    ]["rating"].mean()

    rating_difference = (
        rating_when_mentioned
        - rating_when_not_mentioned
    )

    rows.append({
        "Aspect (ประเด็น)": label,
        "Aspect_Clean": label.split(" (")[0],
        "จำนวนรีวิวเชิงลบที่บ่น (รีวิว)": cnt_neg,
        "สัดส่วนในรีวิวเชิงลบ (%)": pct_neg,
        "สัดส่วนในรีวิวทั้งหมด (%)": pct_all,
        "คะแนนเฉลี่ยเมื่อมีการกล่าวถึง": rating_when_mentioned,
        "คะแนนเฉลี่ยเมื่อไม่มีการกล่าวถึง": rating_when_not_mentioned,
        "ส่วนต่างคะแนนเฉลี่ย": rating_difference
    })

aspect_df = (
    pd.DataFrame(rows)
    .sort_values(
        by="สัดส่วนในรีวิวเชิงลบ (%)",
        ascending=False
    )
    .reset_index(drop=True)
)

display(
    aspect_df[
        [
            "Aspect (ประเด็น)",
            "จำนวนรีวิวเชิงลบที่บ่น (รีวิว)",
            "สัดส่วนในรีวิวเชิงลบ (%)",
            "คะแนนเฉลี่ยเมื่อมีการกล่าวถึง",
            "คะแนนเฉลี่ยเมื่อไม่มีการกล่าวถึง",
            "ส่วนต่างคะแนนเฉลี่ย"
        ]
    ].round(2)
)
````

> **สิ่งที่ได้จาก 2.5.1: Problem Aspect Analysis**
>
> * **1. กลุ่มสินค้าที่ทีมเลือกวิเคราะห์:** `Printers & Accessories` มีรีวิวทั้งหมด **3,942 รีวิว** และเป็นรีวิวเชิงลบ 1–2 ดาวจำนวน **975 รีวิว (24.73%)**
>
> * **2. ประเด็นที่ถูกกล่าวถึงใน Negative Reviews มากที่สุด**
>
>   * **อันดับ 1 — Functionality:** **78.15% (762 จาก 975 รีวิว)**
>   * **อันดับ 2 — Service:** **27.08% (264 จาก 975 รีวิว)**
>   * **อันดับ 3 — Price:** **20.72% (202 จาก 975 รีวิว)**
>
> * **3. ความแตกต่างของคะแนนเฉลี่ย:** รีวิวที่กล่าวถึง Functionality มีคะแนนเฉลี่ยประมาณ **3.59 ดาว** เทียบกับ **4.28 ดาว** ในรีวิวที่ไม่ได้กล่าวถึง Functionality ซึ่งต่างกันประมาณ **-0.69 ดาว** ขณะที่ Service มีส่วนต่างประมาณ **-1.07 ดาว**
>
> * **4. การตีความ:** Functionality เป็นประเด็นที่ถูกกล่าวถึงมากที่สุดใน Negative Reviews ของหมวดนี้ ส่วน Service มีส่วนต่างของคะแนนเฉลี่ยมากที่สุด อย่างไรก็ตาม ส่วนต่างคะแนนดังกล่าวเป็น **ความแตกต่างเชิงสังเกต** และไม่ควรตีความว่า Aspect นั้นเป็นสาเหตุโดยตรงของคะแนนที่ลดลง
>
> * **5. ข้อสรุปสำหรับการตัดสินใจ:** ภายใต้ข้อจำกัดที่ทีมสามารถเลือกประเด็นสำคัญได้ 3 เรื่อง ข้อมูลสนับสนุนให้พิจารณา **Functionality, Service และ Price** เป็น 3 ประเด็นหลักสำหรับการวิเคราะห์และจัดลำดับความสำคัญต่อไป


````python
# ============================================================
# 2.5.2 What-if Simulation
# สมมติฐาน:
# หาก Negative Review ที่เกี่ยวข้องกับ Aspect ที่เลือก
# ได้รับการแก้ไขสำเร็จ ให้ถือว่า Rating เปลี่ยนเป็น 4 ดาว
# ============================================================

base_rating = printers["rating"].mean()
base_negative_rate = printers["negative_review"].mean() * 100

scenarios = [
    (
        "Baseline",
        []
    ),
    (
        "Fix Functionality",
        ["aspect_functionality"]
    ),
    (
        "Fix Functionality + Service",
        ["aspect_functionality", "aspect_service"]
    ),
    (
        "Fix Functionality + Service + Price",
        [
            "aspect_functionality",
            "aspect_service",
            "aspect_price"
        ]
    ),
    (
        "Fix Delivery Only",
        ["aspect_delivery"]
    )
]

simulation_rows = []

for scenario_name, aspect_cols in scenarios:

    # Baseline
    if len(aspect_cols) == 0:

        target_reviews = pd.Series(
            False,
            index=printers.index
        )

    else:

        target_reviews = (
            printers["negative_review"]
            &
            printers[aspect_cols]
            .eq(1)
            .any(axis=1)
        )

    # Copy rating เดิม
    simulated_rating = printers["rating"].copy()

    # สมมติว่าปัญหาได้รับการแก้สำเร็จ
    # Negative Review ที่เข้าเงื่อนไขเปลี่ยนเป็น 4 ดาว
    simulated_rating.loc[target_reviews] = 4

    new_rating = simulated_rating.mean()

    remaining_negative = (
        printers["negative_review"]
        & ~target_reviews
    ).sum()

    new_negative_rate = (
        remaining_negative / len(printers)
    ) * 100

    target_count = target_reviews.sum()

    negative_reduction = (
        target_count / printers["negative_review"].sum()
    ) * 100

    simulation_rows.append({
        "Scenario": scenario_name,
        "จำนวน Negative Reviews ที่จำลองว่าแก้ได้": target_count,
        "คะแนนเฉลี่ยหลัง Simulation": new_rating,
        "Rating Uplift": new_rating - base_rating,
        "Negative Review Rate (%)": new_negative_rate,
        "Negative Reviews ที่แก้ได้ (%)": negative_reduction
    })

simulation_df = pd.DataFrame(
    simulation_rows
)

display(
    simulation_df.round(3)
)
````

>   * ภายในหมวดนี้ `Inkjet Ink Cartridges` มีรีวิวเชิงลบมากที่สุดจำนวน **513 รีวิว** หรือ **52.6% ของ Negative Reviews ในหมวด**
> * **2. ถ้างบประมาณจำกัดและเลือกได้ 3 เรื่อง**
>
>   * จาก Problem Aspect Analysis ประเด็นที่ถูกกล่าวถึงมากที่สุด 3 อันดับแรกคือ
>
>     1. **Functionality — 78.15%**
>     2. **Service — 27.08%**
>     3. **Price — 20.72%**
> * **3. What-if Simulation**
>
>   * ภายใต้สมมติฐานว่า Negative Reviews ที่เกี่ยวข้องกับปัญหาซึ่งได้รับการแก้ไขจะเปลี่ยนเป็น **4 ดาว**
>   * การแก้ **Functionality** เพียงอย่างเดียวทำให้คะแนนเฉลี่ยใน Simulation เพิ่มจากประมาณ **3.85 เป็น 4.39 ดาว**
>   * การแก้ **Functionality + Service** ทำให้คะแนนเพิ่มเป็นประมาณ **4.42 ดาว**
>   * การแก้ครบ **Functionality + Service + Price** ทำให้คะแนนเพิ่มเป็นประมาณ **4.44 ดาว**
> * **4. Counter-Scenario**
>
>   * หากจำลองการแก้ **Delivery เพียงอย่างเดียว** คะแนนเฉลี่ยเพิ่มจากประมาณ 3.85 เป็นเพียง **3.90 ดาว** และ Negative Review Rate ลดจาก 24.73% เหลือประมาณ **22.83%**

### 2.6 Reality Check

ก่อนเอา insight ไปบอกลูกค้า ให้ลองหักล้างมันด้วยตัวเองก่อน อย่างน้อย 1 insight ต้องผ่านขั้นนี้

คำอธิบายทางเลือกที่พบบ่อย
* **ผลของขนาด:** กลุ่มที่ดูโดดเด่นอาจมีตัวอย่างน้อยมาก ตรวจ n ของทุกกลุ่ม
* **ผลของส่วนผสม:** ความต่างอาจมาจากสัดส่วนสินค้าในแต่ละกลุ่มไม่เหมือนกัน ไม่ใช่จากพฤติกรรมลูกค้า ลองคุมให้เทียบกันเฉพาะสินค้าประเภทเดียวกัน
* **ผลของเวลา:** ของเก่ามีเวลาสะสม vote และรีวิวมากกว่า
* **ผลของความยาว:** ข้อความยาวมีโอกาสมีคำที่เราค้นหามากกว่าโดยอัตโนมัติ
* **ผลของคนไม่กี่คน:** ผู้ใช้หรือสินค้าไม่กี่รายอาจครองสัดส่วนใหญ่ของกลุ่ม

เลือกอย่างน้อยหนึ่งคำอธิบาย แล้ว **ทดสอบด้วยตัวเลขจริง** สรุปว่า insight ยังยืนอยู่ได้ หรือต้องปรับถ้อยคำให้แคบลง

````python
# ============================================================
# 2.6 Reality Check
# ทดสอบว่า Insight ยังอยู่หรือไม่เมื่อพิจารณาคำอธิบายทางเลือก
# ============================================================

printers = df[
    df["product_category"] == "Printers & Accessories"
].copy()

total_p = len(printers)

printers["negative_review"] = (
    printers["rating"] <= 2
)

n_neg = printers["negative_review"].sum()

# ============================================================
# 1. Few Items Dominance Effect
# Leave-Out Top 3 ASINs
# ============================================================

asin_counts = (
    printers
    .groupby("parent_asin")
    .agg(
        total=("rating", "count"),
        neg=("negative_review", "sum")
    )
    .sort_values(
        by="neg",
        ascending=False
    )
)

top3_asins = asin_counts.head(3)

top3_neg_cnt = top3_asins["neg"].sum()

top3_neg_share = (
    top3_neg_cnt / n_neg
) * 100

printers_without_top3 = printers[
    ~printers["parent_asin"].isin(
        top3_asins.index
    )
].copy()

n_neg_without_top3 = (
    printers_without_top3["negative_review"]
    .sum()
)

total_without_top3 = len(
    printers_without_top3
)

neg_rate_all = (
    n_neg / total_p
) * 100

neg_rate_without = (
    n_neg_without_top3
    / total_without_top3
) * 100

func_mention_all = (
    printers.loc[
        printers["rating"] <= 2,
        "aspect_functionality"
    ].mean()
) * 100

func_mention_without = (
    printers_without_top3.loc[
        printers_without_top3["rating"] <= 2,
        "aspect_functionality"
    ].mean()
) * 100

reality_df = pd.DataFrame([
    {
        "กลุ่มข้อมูล": "ทั้งหมด",
        "ASINs": len(asin_counts),
        "รีวิวทั้งหมด": total_p,
        "Negative Reviews": n_neg,
        "Negative Rate (%)": neg_rate_all,
        "Functionality Mention Rate (%)":
            func_mention_all
    },
    {
        "กลุ่มข้อมูล": "ตัด Top 3 ASINs",
        "ASINs": len(asin_counts) - 3,
        "รีวิวทั้งหมด": total_without_top3,
        "Negative Reviews": n_neg_without_top3,
        "Negative Rate (%)":
            neg_rate_without,
        "Functionality Mention Rate (%)":
            func_mention_without
    },
    {
        "กลุ่มข้อมูล": "Delta",
        "ASINs": -3,
        "รีวิวทั้งหมด":
            total_without_top3 - total_p,
        "Negative Reviews":
            n_neg_without_top3 - n_neg,
        "Negative Rate (%)":
            neg_rate_without - neg_rate_all,
        "Functionality Mention Rate (%)":
            func_mention_without - func_mention_all
    }
])

print(
    "1. Leave-Out Sensitivity Test"
)

display(
    reality_df.round(2)
)

# ============================================================
# 2. Length Effect
# ใช้ pd.cut เพราะเรากำหนดช่วง <18, 18-47, 48+
# ============================================================

printers["word_count"] = (
    printers["text"]
    .fillna("")
    .astype(str)
    .apply(lambda x: len(x.split()))
)

printers["length_bucket"] = pd.cut(
    printers["word_count"],
    bins=[-1, 17, 47, np.inf],
    labels=[
        "Short (<18 words)",
        "Medium (18-47 words)",
        "Long (48+ words)"
    ]
)

length_rows = []

for bucket, group in printers.groupby(
    "length_bucket",
    observed=False
):

    neg_group = group[
        group["rating"] <= 2
    ]

    pos_group = group[
        group["rating"] >= 4
    ]

    func_neg = (
        neg_group["aspect_functionality"]
        .mean()
    ) * 100

    func_pos = (
        pos_group["aspect_functionality"]
        .mean()
    ) * 100

    length_rows.append({
        "ช่วงความยาวรีวิว": bucket,
        "Negative n": len(neg_group),
        "Negative Functionality (%)":
            func_neg,
        "Positive n": len(pos_group),
        "Positive Functionality (%)":
            func_pos,
        "Difference (pp)":
            func_neg - func_pos
    })

length_df = pd.DataFrame(
    length_rows
)

print(
    "\n2. Controlled Length Buckets"
)

display(
    length_df.round(2)
)
````
