## เทคนิคที่อาจใช้ — เลือกเท่าที่จำเป็น

ไม่ต้องใช้ทั้งหมด โดยทั่วไป **2–4 เทคนิคก็เพียงพอ** ให้เลือกตามคำถามของทีม

| เทคนิค | เหมาะเมื่ออยากรู้ |
|---|---|
| **Word / n-gram frequency** | คนพูดถึงอะไรบ่อย |
| **TF-IDF / distinctive words** | สองกลุ่มพูดต่างกันอย่างไร |

| **Topic modeling** | มีประเด็นอะไรที่เราอาจยังไม่รู้ |
| **Trend / segmentation** | ปัญหาเปลี่ยนตามเวลา หรือหนักในกลุ่มไหน |

> เทคนิคขั้นสูง เช่น embedding, clustering, regression, classification หรือ anomaly detection ทำได้ แต่ **ไม่บังคับ** และเหมาะสำหรับ Bonus เมื่อใช้ถูกและอธิบายได้

ถ้ายังไม่เคยใช้เทคนิคใด ให้ทดลองกับข้อมูลชุดเล็กก่อน แล้วตรวจด้วยตาว่าผลสมเหตุสมผลหรือไม่

```python
path = "/content/posts_20241127_075630.jsonl"

records = []

with open(path, "r", encoding="utf-8") as f:
    for line in f:
        line = line.strip()
        if line:
            records.append(json.loads(line))

df = pd.DataFrame(records)

print("จำนวนโพสต์:", len(df))
print("Columns:")
print(df.columns.tolist())

display(df.head())
```

```python
print("จำนวนแถว:", len(df))
print("จำนวนคอลัมน์:", len(df.columns))

display(
    df.isnull().sum().sort_values(ascending=False)
)
```

```python
df["created_at"] = pd.to_datetime(
    df["created_at"],
    errors="coerce",
    utc=True
)

df_time = df.dropna(subset=["created_at"]).copy()
start_time = df_time["created_at"].min()
end_time = df_time["created_at"].max()

print("========== Q1: เวลาโพสต์ ==========")
print("ข้อมูลเริ่มต้น :", start_time)
print("ข้อมูลสิ้นสุด :", end_time)
df_time["hour"] = df_time["created_at"].dt.hour

hourly_posts = (
    df_time["hour"]
    .value_counts()
    .sort_index()
)
peak_hour = hourly_posts.idxmax()
peak_count = hourly_posts.max()

print("\nชั่วโมงที่โพสต์มากที่สุด:", peak_hour, "นาฬิกา")
print("จำนวนโพสต์:", peak_count)

# กราฟ
display(
    hourly_posts.rename("posts").to_frame()
)
plt.figure(figsize=(12, 5))

plt.bar(
    hourly_posts.index,
    hourly_posts.values
)

plt.xlabel("Hour")
plt.ylabel("Number of Posts")
plt.title("Bluesky Posts by Hour")

plt.xticks(range(24))
plt.tight_layout()
plt.show()
```

```python
# ==========================================
# Q3: คนพูดถึงอะไร?
# ==========================================
# ------------------------------------------
# 1. หา Hashtag
# ------------------------------------------
from collections import Counter
import re
import pandas as pd

hashtags = []

for text in df["text"].dropna():

    found = re.findall(
        r'#[\w]+',
        str(text)
    )

    hashtags.extend(
        [h.lower() for h in found]
    )

# นับความถี่
hashtag_counts = Counter(hashtags)

# สร้างตาราง Top 20
hashtag_df = pd.DataFrame(
    hashtag_counts.most_common(20),
    columns=["hashtag", "count"]
)

print("========== TOP 20 HASHTAGS ==========")

display(hashtag_df)
```

```python
# ------------------------------------------
# 2. หา Keyword
# ------------------------------------------

stopwords = {
    "the", "and", "for", "that", "this",
    "with", "you", "are", "was", "have",
    "from", "they", "but", "not", "what",
    "just", "about", "your", "all",
    "will", "can", "has", "had", "its",
    "his", "her", "our", "out", "who",
    "get", "got", "one", "more", "too",
    "how", "why", "when", "where",
    "been", "being", "than", "then",
    "http", "https", "www", "com"
}

words = []

for text in df["text"].dropna():

    # เอาเฉพาะคำภาษาอังกฤษ
    tokens = re.findall(
        r'\b[a-zA-Z]{3,}\b',
        str(text).lower()
    )

    for word in tokens:

        if word not in stopwords:
            words.append(word)

# นับความถี่
word_counts = Counter(words)

# Top 20
keyword_df = pd.DataFrame(
    word_counts.most_common(20),
    columns=["keyword", "count"]
)

print("========== TOP 20 KEYWORDS ==========")

display(keyword_df)
```

```python
# กราฟ keywords
top_keywords = keyword_df.head(10).sort_values(
    "count",
    ascending=True
)

plt.figure(figsize=(10, 6))

plt.barh(
    top_keywords["keyword"],
    top_keywords["count"]
)

plt.xlabel("Frequency")
plt.ylabel("Keyword")
plt.title("Top 10 Keywords in Bluesky Posts")

plt.tight_layout()
plt.show()
```

```python
# ------------------------------------------
# 3. ตัวอย่างข้อความจริง
# ------------------------------------------

top_keyword = keyword_df.iloc[0]["keyword"]

print("Keyword ที่เลือก:", top_keyword)
print("=" * 80)

examples = df[
    df["text"]
    .fillna("")
    .str.contains(
        top_keyword,
        case=False,
        na=False
    )
][
    ["text", "created_at"]
].head(5)

display(examples)
```

```python
# เลือก Keyword ที่ต้องการ
selected_keyword = "Downe"

examples = df[
    df["text"]
    .fillna("")
    .str.contains(
        selected_keyword,
        case=False,
        na=False
    )
][
    ["text", "created_at"]
].head(10)

print("Keyword:", selected_keyword)
print("จำนวนตัวอย่าง:", len(examples))

display(examples)
```

```python
selected_hashtag = hashtag_df.iloc[0]["hashtag"]

examples_hashtag = df[
    df["text"]
    .fillna("")
    .str.contains(
        re.escape(selected_hashtag),
        case=False,
        na=False
    )
][
    ["text", "created_at"]
].head(5)

print("Hashtag:", selected_hashtag)
print("=" * 80)

display(examples_hashtag)
```

```python
# ==========================================
# Q4: Visualization - จำนวนโพสต์ตามปี
# ==========================================

import pandas as pd
import matplotlib.pyplot as plt

# แปลงวันที่
df["created_at"] = pd.to_datetime(
    df["created_at"],
    errors="coerce",
    utc=True
)

# เอาเฉพาะข้อมูลที่มีวันที่
df_time = df.dropna(subset=["created_at"]).copy()

# สร้างปี
df_time["year"] = df_time["created_at"].dt.year

# นับจำนวนโพสต์แต่ละปี
yearly_posts = (
    df_time
    .groupby("year")
    .size()
    .reset_index(name="posts")
)

print("จำนวนโพสต์แยกตามปี")
display(yearly_posts)
```

```python
plt.figure(figsize=(12, 6))

plt.bar(
    yearly_posts["year"].astype(str),
    yearly_posts["posts"]
)

plt.yscale("log")

plt.xlabel("Year")
plt.ylabel("Number of Posts (log scale)")
plt.title("Number of Bluesky Posts by Year")

plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```
